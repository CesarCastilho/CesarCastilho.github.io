# 4. Equações Homogêneas

Nesta seção estudaremos uma classe de EDOs de primeira ordem que **pode ser reduzida a uma equação com variáveis separáveis** 
por uma mudança de variável.

A palavra **homogênea** aqui se refere à homogeneidade de funções de duas variáveis. Ela não deve ser confundida com o uso posterior da expressão **equação linear homogênea**.

A ideia central será reconhecer equações cuja inclinação depende apenas da razão $y/x$. Geometricamente, isso significa que a inclinação é constante ao longo de cada reta que passa pela origem.

## 4.1 Funções homogêneas

Uma função $F(x,y)$ é **homogênea de grau $m$** se

$$
F(t \, x,t \, y)=t^m \, F(x,y).
$$

Por exemplo,

$$
F(x,y)=x^2+x \, y+y^2
$$

é homogênea de grau $2$. Já

$$
G(x,y)=\frac{x^2-y^2}{x^2+y^2}
$$

é homogênea de grau zero.

As funções homogêneas de grau zero serão particularmente importantes.

## 4.2 Grau zero e a razão $y/x$

Suponha que $F$ seja homogênea de grau zero:

$$
F(tx,ty)=F(x,y).
$$

Se $x\neq0$, tomando $t=1/x$,

$$
F(x,y)=F\left(1,\frac{y}{x}\right).
$$

Definindo $\phi(u)=F(1,u)$,

$$
\boxed{F(x,y)=\phi\left(\frac{y}{x}\right)}.
$$

Assim, uma função homogênea de grau zero depende essencialmente apenas da razão $y/x$.

## 4.3 Equações diferenciais homogêneas

Uma EDO

$$
y'=F(x,y)
$$

é chamada **homogênea** quando $F$ é homogênea de grau zero. Podemos então escrevê-la como

$$
y'=\phi\left(\frac{y}{x}\right).
$$

A substituição natural é

$$
u=\frac{y}{x},
\qquad y=xu.
$$

Como $u=u(x)$,

$$
y'=u+xu'.
$$

Substituindo,

$$
u+xu'=\phi(u),
$$

logo

$$
\boxed{xu'=\phi(u)-u}.
$$

Quando $\phi(u)-u\neq0$,

$$
\frac{du}{\phi(u)-u}=\frac{dx}{x},
$$

que é separável.

## 4.4 Antes de dividir

Devemos primeiro resolver

$$
\phi(u)-u=0.
$$

Se $u_*$ é uma raiz, então $u(x)=u_*$ é solução da equação transformada e

$$
\boxed{y=u_*x}
$$

é uma possível solução da EDO original.

Essas soluções podem ser perdidas se dividirmos imediatamente por $\phi(u)-u$.

## 4.5 A intuição geométrica

Ao longo da reta

$$
y=kx,
$$

temos $y/x=k$. Portanto,

$$
y'=\phi(k)
$$

em todos os seus pontos: cada reta pela origem é uma **isóclina**.

Quando essa isóclina é também uma curva integral?

A própria reta $y=kx$ possui inclinação $k$, enquanto o campo sobre ela possui inclinação $\phi(k)$. Logo, ela é solução exatamente quando

$$
\boxed{k=\phi(k)}.
$$

Esta é precisamente a equação $\phi(u)-u=0$ que surgiu na substituição.

Assim, a condição algébrica tem uma interpretação geométrica direta: **o campo de direções deve ser tangente à própria isóclina**.

## 4.6 Exemplo: $y'=(x+y)/(x-y)$

Dividindo numerador e denominador por $x$,

$$
y'=\frac{1+y/x}{1-y/x}.
$$

Logo,

$$
\phi(u)=\frac{1+u}{1-u},
$$

e a equação é homogênea.

Vamos visualizar a propriedade característica: ao longo de cada reta $y=kx$, a inclinação do campo permanece constante.

```python
import numpy as np
import matplotlib.pyplot as plt

def campo_direcoes(f, xlim=(-4,4), ylim=(-4,4), nx=25, ny=25):
    xs=np.linspace(xlim[0],xlim[1],nx)
    ys=np.linspace(ylim[0],ylim[1],ny)
    X,Y=np.meshgrid(xs,ys)
    with np.errstate(divide='ignore',invalid='ignore'):
        M=f(X,Y)
    U=1/np.sqrt(1+M**2)
    V=M/np.sqrt(1+M**2)
    mask=np.isfinite(M)
    plt.figure(figsize=(8,7))
    plt.quiver(X[mask],Y[mask],U[mask],V[mask],
               angles='xy',pivot='mid',
               headwidth=0,headlength=0,headaxislength=0)
    plt.axhline(0,linewidth=.8); plt.axvline(0,linewidth=.8)
    plt.xlim(xlim); plt.ylim(ylim)
    plt.xlabel("x"); plt.ylabel("y"); plt.grid(alpha=.2)

f=lambda x,y:(x+y)/(x-y)
campo_direcoes(f)
x1=np.linspace(-4,-.05,300); x2=np.linspace(.05,4,300)
for k in [-2,-1,-.5,0,.5,2]:
    plt.plot(x1,k*x1,linewidth=1)
    plt.plot(x2,k*x2,linewidth=1)
plt.title(r"Campo de direções de $y'=(x+y)/(x-y)$")
plt.show()
```

## 4.7 Resolvendo o exemplo

Faça $y=xu$. Então

$$
u+xu'=\frac{1+u}{1-u}.
$$

Logo,

$$
xu'=\frac{1+u^2}{1-u},
$$

e

$$
\frac{1-u}{1+u^2}\,du=\frac{dx}{x}.
$$

Integrando,

$$
\arctan u-\frac12\ln(1+u^2)=\ln|x|+C.
$$

Substituindo $u=y/x$ e simplificando,

$$
\boxed{
\arctan\left(\frac{y}{x}\right)
-\frac12\ln(x^2+y^2)=C.
}
$$

A forma implícita é natural neste problema.

## 4.8 Existem retas que são soluções?

Uma reta $y=kx$ seria solução se

$$
k=\frac{1+k}{1-k}.
$$

Isso fornece

$$
-k^2=1,
$$

sem raízes reais.

Portanto, todas as retas $y=kx$ são isóclinas, mas nenhuma delas é uma curva integral real neste exemplo.

## 4.9 Um exemplo com retas que são soluções

Considere

$$
y'=\frac{x^2+y^2}{2xy}.
$$

Temos

$$
\phi(u)=\frac{1+u^2}{2u}.
$$

Uma reta $y=kx$ é solução quando

$$
k=\frac{1+k^2}{2k},
$$

isto é,

$$
k^2=1.
$$

Logo,

$$
\boxed{y=x}
\qquad\text{e}\qquad
\boxed{y=-x}
$$

são curvas integrais.

```python
f2=lambda x,y:(x**2+y**2)/(2*x*y)
campo_direcoes(f2,(-4,4),(-4,4),27,27)
x1=np.linspace(-4,-.08,300); x2=np.linspace(.08,4,300)
plt.plot(x1,x1,linewidth=2.5,label=r"$y=x$")
plt.plot(x2,x2,linewidth=2.5)
plt.plot(x1,-x1,linewidth=2.5,label=r"$y=-x$")
plt.plot(x2,-x2,linewidth=2.5)
plt.title(r"Campo de direções de $y'=(x^2+y^2)/(2xy)$")
plt.legend(); plt.show()
```

## 4.10 Resolvendo a equação

Com $y=xu$,

$$
u+xu'=\frac{1+u^2}{2u},
$$

portanto

$$
xu'=\frac{1-u^2}{2u}.
$$

Antes de dividir, $1-u^2=0$ recupera $u=\pm1$, isto é, $y=\pm x$.

Para as demais soluções,

$$
\frac{2u}{1-u^2}\,du=\frac{dx}{x}.
$$

Integrando,

$$
-\ln|1-u^2|=\ln|x|+C.
$$

Após substituir $u=y/x$ e reorganizar,

$$
\boxed{x^2-y^2=Cx}.
$$

Essa é a família das soluções não excepcionais, em intervalos compatíveis com o domínio da EDO.

## 4.11 Visualização da família implícita

Vamos visualizar

$$
x^2-y^2=Cx
$$

sem isolar $y$.

```python
x=np.linspace(-4,4,600); y=np.linspace(-4,4,600)
X,Y=np.meshgrid(x,y)
plt.figure(figsize=(8,7))
for C in [-4,-2,-1,1,2,4]:
    plt.contour(X,Y,X**2-Y**2-C*X,levels=[0],linewidths=1.5)
plt.plot(x,x,linewidth=2.2,label=r"$y=x$")
plt.plot(x,-x,linewidth=2.2,label=r"$y=-x$")
plt.axhline(0,linewidth=.8); plt.axvline(0,linewidth=.8)
plt.xlim(-4,4); plt.ylim(-4,4)
plt.xlabel("x"); plt.ylabel("y")
plt.title(r"Curvas $x^2-y^2=Cx$ e soluções $y=\pm x$")
plt.grid(alpha=.2); plt.legend(); plt.show()
```

## 4.12 Forma diferencial

Uma equação

$$
P(x,y)\,dx+Q(x,y)\,dy=0
$$

é homogênea quando $P$ e $Q$ são funções homogêneas do **mesmo grau**.

Quando $Q\neq0$,

$$
y'=-\frac{P(x,y)}{Q(x,y)}.
$$

Se ambas têm grau $m$,

$$
\frac{P(tx,ty)}{Q(tx,ty)}
=
\frac{t^mP(x,y)}{t^mQ(x,y)}
=
\frac{P(x,y)}{Q(x,y)},
$$

portanto $-P/Q$ tem grau zero e podemos usar $y=xu$.

## 4.13 A invariância por mudança de escala

Se

$$
F(tx,ty)=F(x,y),
$$

os pontos $(x,y)$ e $(tx,ty)$ possuem a mesma inclinação no campo. Eles pertencem à mesma reta pela origem.

Assim, a transformação

$$
(x,y)\longmapsto(tx,ty)
$$

preserva o padrão de direções.

A razão

$$
\frac{y}{x}
$$

também é preservada:

$$
\frac{ty}{tx}=\frac{y}{x}.
$$

Ou seja, $u=y/x$  variável natural associada à **simetria de escala** da equação.

## 4.14 Equações transformáveis em homogêneas

Há equações que não são homogêneas na forma em que aparecem originalmente, mas que podem ser transformadas em equações homogêneas por meio de uma **translação das coordenadas**.

Considere uma EDO da forma

$$
y'
=
F\left(
\frac{a_1x+b_1y+c_1}
     {a_2x+b_2y+c_2}
\right).
$$

A presença dos termos constantes $c_1$ e $c_2$ impede, em geral, que o lado direito dependa apenas da razão $y/x$.

A ideia será escolher uma nova origem para o sistema de coordenadas de modo que esses dois termos constantes desapareçam simultaneamente.

### Por que consideramos a interseção de duas retas?

Considere as retas

$$
a_1x+b_1y+c_1=0
$$

e

$$
a_2x+b_2y+c_2=0.
$$

Suponha que elas se intersectem em um ponto $(x_0,y_0)$.

Isso significa que $(x_0,y_0)$ satisfaz simultaneamente

$$
a_1x_0+b_1y_0+c_1=0
$$

e

$$
a_2x_0+b_2y_0+c_2=0.
$$

É exatamente essa propriedade que precisamos para eliminar os termos constantes.

Vamos fazer a translação

$$
x=X+x_0,
\qquad
y=Y+y_0,
$$

ou, equivalentemente,

$$
X=x-x_0,
\qquad
Y=y-y_0.
$$

Geometricamente, estamos simplesmente escolhendo o ponto $(x_0,y_0)$ como a **nova origem** do sistema de coordenadas.

### O que acontece com o numerador?

Substituindo

$$
x=X+x_0,
\qquad
y=Y+y_0,
$$

temos

$$
a_1x+b_1y+c_1
=
a_1(X+x_0)+b_1(Y+y_0)+c_1.
$$

Reorganizando,

$$
a_1x+b_1y+c_1
=
a_1X+b_1Y+
\left(a_1x_0+b_1y_0+c_1\right).
$$

Mas $(x_0,y_0)$ pertence à primeira reta. Portanto,

$$
a_1x_0+b_1y_0+c_1=0.
$$

Logo,

$$
\boxed{
a_1x+b_1y+c_1=a_1X+b_1Y.
}
$$

O termo constante desapareceu.

### E o denominador?

Da mesma forma,

$$
a_2x+b_2y+c_2
=
a_2(X+x_0)+b_2(Y+y_0)+c_2,
$$

e portanto

$$
a_2x+b_2y+c_2
=
a_2X+b_2Y+
\left(a_2x_0+b_2y_0+c_2\right).
$$

Como $(x_0,y_0)$ também pertence à segunda reta,

$$
a_2x_0+b_2y_0+c_2=0.
$$

Assim,

$$
\boxed{
a_2x+b_2y+c_2=a_2X+b_2Y.
}
$$

Portanto, a mesma translação eliminou **simultaneamente** os dois termos constantes.

### Por que isso transforma a equação em uma equação homogênea?

Uma translação não altera a derivada, pois

$$
y=Y+y_0
$$

e

$$
x=X+x_0,
$$

com $x_0$ e $y_0$ constantes. Assim,

$$
\frac{dy}{dx}
=
\frac{dY}{dX}.
$$

A EDO transforma-se em

$$
\frac{dY}{dX}
=
F\left(
\frac{a_1X+b_1Y}
     {a_2X+b_2Y}
\right).
$$

Agora dividimos numerador e denominador por $X$:

$$
\frac{a_1X+b_1Y}
     {a_2X+b_2Y}
=
\frac{
a_1+b_1\frac{Y}{X}
}{
a_2+b_2\frac{Y}{X}
}.
$$

Portanto,

$$
\frac{dY}{dX}
=
F\left(
\frac{
a_1+b_1\frac{Y}{X}
}{
a_2+b_2\frac{Y}{X}
}
\right).
$$

Se definirmos

$$
\phi(u)
=
F\left(
\frac{a_1+b_1u}
     {a_2+b_2u}
\right),
$$

obtemos

$$
\boxed{
\frac{dY}{dX}
=
\phi\left(\frac{Y}{X}\right).
}
$$

Esta é precisamente uma equação homogênea de primeira ordem.

Podemos então utilizar a substituição usual

$$
u=\frac{Y}{X},
\qquad
Y=uX.
$$

### A interpretação geométrica

A equação original não apresenta a estrutura homogênea em relação à origem $(0,0)$ porque os termos $c_1$ e $c_2$ deslocam essa estrutura.

O ponto de interseção $(x_0,y_0)$ desempenha um papel especial: ele é o único ponto que permite anular simultaneamente os termos constantes do numerador e do denominador.

Ao fazer

$$
X=x-x_0,
\qquad
Y=y-y_0,
$$

estamos colocando a nova origem exatamente nesse ponto.

Em relação às novas coordenadas $(X,Y)$, a equação passa a depender apenas da razão

$$
\frac{Y}{X},
$$

e recuperamos a estrutura radial característica das equações homogêneas.

Assim, o ponto $(x_0,y_0)$ não aparece por acaso: **ele é precisamente a origem natural da estrutura homogênea escondida na equação original.**

### Quando existe esse ponto de interseção?

Para determinar $(x_0,y_0)$ precisamos resolver o sistema

$$
\begin{cases}
a_1x_0+b_1y_0=-c_1,\\
a_2x_0+b_2y_0=-c_2.
\end{cases}
$$

Esse sistema possui uma única solução quando

$$
\boxed{
a_1b_2-a_2b_1\neq 0.
}
$$

Geometricamente, essa condição significa que as duas retas não são paralelas e, portanto, possuem um único ponto de interseção.

Se as retas forem paralelas e distintas, não existe um ponto $(x_0,y_0)$ que pertença simultaneamente às duas. Nesse caso, não é possível escolher uma translação que elimine ao mesmo tempo os dois termos constantes, e **esse procedimento específico de redução a uma equação homogênea não se aplica**.

> **Ideia central:** procuramos o ponto de interseção das duas retas porque queremos transformá-lo na nova origem. Nesse ponto, tanto o numerador quanto o denominador perdem seus termos constantes, revelando a estrutura homogênea da equação.

### Exemplo

Considere

$$
y'=\frac{x+y-1}{x-y+3}.
$$

As retas do numerador e denominador intersectam-se em $(-1,2)$.

Faça

$$
x=X-1,\qquad y=Y+2.
$$

Então

$$
x+y-1=X+Y,
$$

$$
x-y+3=X-Y,
$$

e a equação torna-se

$$
\boxed{\frac{dY}{dX}=\frac{X+Y}{X-Y}},
$$

que é homogênea. A partir daqui usamos $Y=uX$.

## 4.15 Quando a translação não funciona dessa forma?

O procedimento depende de

$$
\Delta=
\begin{vmatrix}
a_1&b_1\\
a_2&b_2
\end{vmatrix}
\neq0.
$$

Se $\Delta=0$, as retas são paralelas ou coincidentes e não há, em geral, um único ponto que elimine simultaneamente os termos constantes.

Nesses casos, outra mudança de variável pode ser mais adequada, frequentemente uma combinação linear como $z=ax+by$.

Reconhecer a estrutura continua sendo mais importante do que aplicar mecanicamente uma substituição.

## 4.16 Roteiro

Para uma EDO homogênea:

1. Verifique se $F(tx,ty)=F(x,y)$, ou escreva $F(x,y)=\phi(y/x)$.
2. Faça $u=y/x$, isto é, $y=xu$.
3. Use $y'=u+xu'$.
4. Obtenha $xu'=\phi(u)-u$.
5. **Antes de dividir**, resolva $\phi(u)-u=0$; as raízes podem fornecer $y=u_*x$.
6. Para as demais soluções, separe:
$$
\frac{du}{\phi(u)-u}=\frac{dx}{x}.
$$
7. Integre e retorne a $u=y/x$.
8. Examine cuidadosamente o domínio da equação original e da solução.
