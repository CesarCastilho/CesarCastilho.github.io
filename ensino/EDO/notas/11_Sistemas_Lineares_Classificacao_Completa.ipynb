## 1. Retomando a aula anterior

Na aula anterior vimos que, se

$$
A\mathbf v=\lambda\mathbf v,
$$

então a reta gerada por $\mathbf v$ é invariante pelo sistema

$$
\mathbf{x}'=A\mathbf{x}.
$$

Sobre essa reta, uma solução possui a forma

$$
\mathbf{x}(t)=Ce^{\lambda t}\mathbf v.
$$

Se $A$ possui dois autovetores reais linearmente independentes,

$$
A\mathbf v_1=\lambda_1\mathbf v_1,
\qquad
A\mathbf v_2=\lambda_2\mathbf v_2,
$$

então qualquer condição inicial pode ser escrita como

$$
\mathbf{x}_0=c_1\mathbf v_1+c_2\mathbf v_2,
$$

e a solução correspondente é

$$
\boxed{
\mathbf{x}(t)
=
c_1e^{\lambda_1t}\mathbf v_1
+
c_2e^{\lambda_2t}\mathbf v_2.
}
$$

Queremos  **ler a geometria do sistema diretamente a partir dela**.

# Parte I — Dois autovalores reais

## 2. Duas direções que decaem

Considere inicialmente

$$
\begin{cases}
x'=-x,\\
y'=-3y.
\end{cases}
$$

ou, matricialmente,

$$
\mathbf{x}'
=
\begin{pmatrix}
-1&0\\
0&-3
\end{pmatrix}
\mathbf{x}.
$$

Antes de resolver analiticamente, vamos observar o campo vetorial e algumas trajetórias.

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[-1.0, 0.0],
              [ 0.0,-3.0]])

x = np.linspace(-3, 3, 25)
y = np.linspace(-3, 3, 25)
X, Y = np.meshgrid(x, y)

U = A[0,0]*X + A[0,1]*Y
V = A[1,0]*X + A[1,1]*Y

speed = np.sqrt(U**2 + V**2)
speed[speed == 0] = 1

plt.figure(figsize=(7,7))
plt.streamplot(X, Y, U, V, density=1.1, linewidth=1)

for x0, y0 in [(2,2), (2,-2), (-2,2), (-2,-2), (2,0.7), (-1.5,-2.5)]:
    t = np.linspace(0, 4, 300)
    xt = x0*np.exp(-t)
    yt = y0*np.exp(-3*t)
    plt.plot(xt, yt)

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$x'=-x,\quad y'=-3y$")
plt.grid(alpha=0.2)
plt.show()
```

### Responda

1. Para onde caminham as trajetórias quando $t\to+\infty$?
2. Existe alguma direção especial?
3. Qual componente desaparece mais rapidamente?
4. Quando uma trajetória se aproxima da origem, ela tende a ficar paralela a qual eixo?
5. O que esperamos que os autovalores nos digam sobre essas observações?



## 3. Solução analítica

As equações estão desacopladas:

$$
x'=-x,
\qquad
y'=-3y.
$$

Logo,

$$
x(t)=c_1e^{-t},
\qquad
y(t)=c_2e^{-3t}.
$$

Os autovalores são

$$
\lambda_1=-1,
\qquad
\lambda_2=-3,
$$

com autovetores

$$
\mathbf e_1=
\begin{pmatrix}1\\0\end{pmatrix},
\qquad
\mathbf e_2=
\begin{pmatrix}0\\1\end{pmatrix}.
$$

Como ambos os autovalores são negativos,

$$
e^{-t}\to0,
\qquad
e^{-3t}\to0
\qquad
(t\to+\infty).
$$

Portanto,

$$
\boxed{\mathbf{x}(t)\to\mathbf 0.}
$$

A origem atrai todas as soluções.

Mas os dois termos não desaparecem com a mesma velocidade:

$$
e^{-3t}\ll e^{-t}
$$

para $t$ grande.

Assim, genericamente,

$$
\mathbf{x}(t)
=
c_1e^{-t}\mathbf e_1+c_2e^{-3t}\mathbf e_2
$$

fica progressivamente dominada pelo termo associado ao autovalor **menos negativo**, isto é, $\lambda=-1$.

Consequentemente, as trajetórias tendem à origem tangentes à direção de $\mathbf e_1$.

Esse retrato de fase é chamado de **nó estável** ou **sumidouro**.

### Regra geométrica — dois autovalores reais negativos

Se

$$
\lambda_1<0,
\qquad
\lambda_2<0,
\qquad
\lambda_1\neq\lambda_2,
$$

e existem dois autovetores linearmente independentes, então a origem é um

$$
\boxed{\text{nó estável}.}
$$

Todas as trajetórias suficientemente consideradas no plano convergem para a origem quando $t\to+\infty$.

Se

$$
\lambda_1>\lambda_2,
$$

com ambos negativos, então $\lambda_1$ é o autovalor menos negativo e

$$
e^{\lambda_1t}
$$

decai mais lentamente.

Para condições iniciais genéricas, a direção de aproximação à origem é a direção do autovetor associado a $\lambda_1$.

## 4. Dois autovalores positivos.

Considere

$$
\begin{cases}
x'=x,\\
y'=3y.
\end{cases}
$$

A solução é

$$
x(t)=c_1e^t,
\qquad
y(t)=c_2e^{3t}.
$$

Agora,

$$
e^t\to\infty,
\qquad
e^{3t}\to\infty
$$

quando $t\to+\infty$.

Exceto pela solução de equilíbrio, as trajetórias afastam-se da origem.

Temos um

$$
\boxed{\text{nó instável}}
$$

ou **fonte**.

Assim:

$$
\boxed{
\lambda_1,\lambda_2<0
\Rightarrow
\text{nó estável}
}
$$

e

$$
\boxed{
\lambda_1,\lambda_2>0
\Rightarrow
\text{nó instável}.
}
$$

# Parte II — A sela

## 5. O que acontece quando os autovalores têm sinais opostos?

Considere

$$
\begin{cases}
x'=x,\\
y'=-2y.
\end{cases}
$$

Antes do cálculo, observe o retrato de fase.

```python
A = np.array([[1.0, 0.0],
              [0.0,-2.0]])

x = np.linspace(-3, 3, 25)
y = np.linspace(-3, 3, 25)
X, Y = np.meshgrid(x, y)
U = X
V = -2*Y

plt.figure(figsize=(7,7))
plt.streamplot(X, Y, U, V, density=1.2, linewidth=1)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$x'=x,\quad y'=-2y$")
plt.grid(alpha=0.2)
plt.show()
```

A solução é

$$
x(t)=c_1e^t,
\qquad
y(t)=c_2e^{-2t}.
$$

Na direção do eixo $x$,

$$
e^t
$$

cresce.

Na direção do eixo $y$,

$$
e^{-2t}
$$

decai.

Temos, portanto, simultaneamente:

- uma direção de afastamento;
- uma direção de aproximação.

Esse equilíbrio é chamado de **sela**.

$$
\boxed{
\lambda_1\lambda_2<0
\quad\Longrightarrow\quad
\text{sela}.
}
$$

A reta associada ao autovalor negativo é a direção estável.

A reta associada ao autovalor positivo é a direção instável.

## 6. Uma propriedade importante da sela

Considere

$$
\mathbf{x}(t)
=
c_1e^{\lambda_1t}\mathbf v_1
+
c_2e^{\lambda_2t}\mathbf v_2,
$$

com

$$
\lambda_1>0,
\qquad
\lambda_2<0.
$$

Para que

$$
\mathbf{x}(t)\to0
\qquad
(t\to+\infty),
$$

é necessário eliminar o termo crescente:

$$
c_1=0.
$$

Portanto, apenas as condições iniciais sobre

$$
\operatorname{span}(\mathbf v_2)
$$

convergem para a origem no futuro.

Chamamos essa reta de **direção estável**.

Analogamente, a reta

$$
\operatorname{span}(\mathbf v_1)
$$

é a **direção instável**.

A sela é instável porque condições iniciais arbitrariamente próximas da origem, mas fora da direção estável, possuem uma componente que cresce exponencialmente.

# Parte III — Um exemplo não diagonal

## 7. Quais as direções importantes?

Considere

$$
\mathbf{x}'
=
A\mathbf{x},
\qquad
A=
\begin{pmatrix}
1&2\\
2&1
\end{pmatrix}.
$$

Agora as equações são acopladas:

$$
\begin{cases}
x'=x+2y,\\
y'=2x+y.
\end{cases}
$$

Vamos primeiro observar o retrato de fase.

```python
A = np.array([[1.0,2.0],
              [2.0,1.0]])

x = np.linspace(-3,3,30)
y = np.linspace(-3,3,30)
X,Y = np.meshgrid(x,y)
U = A[0,0]*X + A[0,1]*Y
V = A[1,0]*X + A[1,1]*Y

plt.figure(figsize=(7,7))
plt.streamplot(X,Y,U,V,density=1.2)
plt.axhline(0,linewidth=.8)
plt.axvline(0,linewidth=.8)
plt.plot([-3,3],[-3,3],"--",linewidth=1)
plt.plot([-3,3],[3,-3],"--",linewidth=1)
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Retrato de fase antes do cálculo espectral")
plt.grid(alpha=.2)
plt.show()
```

O polinômio característico é

$$
\det(A-\lambda I)
=
\begin{vmatrix}
1-\lambda&2\\
2&1-\lambda
\end{vmatrix}.
$$

Assim,

$$
(1-\lambda)^2-4=0,
$$

de onde

$$
\lambda_1=3,
\qquad
\lambda_2=-1.
$$

Para $\lambda_1=3$, podemos tomar

$$
\mathbf v_1=
\begin{pmatrix}
1\\1
\end{pmatrix}.
$$

Para $\lambda_2=-1$,

$$
\mathbf v_2=
\begin{pmatrix}
1\\-1
\end{pmatrix}.
$$

Portanto,

$$
\boxed{
\mathbf{x}(t)
=
c_1e^{3t}
\begin{pmatrix}1\\1\end{pmatrix}
+
c_2e^{-t}
\begin{pmatrix}1\\-1\end{pmatrix}.
}
$$

Como os autovalores têm sinais opostos, a origem é uma sela.

Mas agora as direções estável e instável não são os eixos coordenados:

$$
\boxed{
y=-x
\quad\text{é a direção estável},
}
$$

enquanto

$$
\boxed{
y=x
\quad\text{é a direção instável}.
}
$$

Isso mostra por que os autovetores são geometricamente fundamentais: eles revelam as direções invariantes naturais do sistema.

# Parte IV — Quando os autovalores deixam de ser reais

## 8. Um sistema que gira

Considere

$$
\begin{cases}
x'=-y,\\
y'=x.
\end{cases}
$$

ou

$$
\mathbf{x}'
=
\begin{pmatrix}
0&-1\\
1&0
\end{pmatrix}
\mathbf{x}.
$$

Observe numericamente o campo.

```python
x = np.linspace(-3,3,25)
y = np.linspace(-3,3,25)
X,Y = np.meshgrid(x,y)

U = -Y
V = X

plt.figure(figsize=(7,7))
plt.streamplot(X,Y,U,V,density=1.2)
plt.axhline(0,linewidth=.8)
plt.axvline(0,linewidth=.8)
plt.axis("equal")
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$x'=-y,\quad y'=x$")
plt.grid(alpha=.2)
plt.show()
```

O polinômio característico é

$$
\lambda^2+1=0.
$$

Logo,

$$
\lambda=\pm i.
$$

Não existem autovetores reais associados a esses autovalores.

Entretanto, o sistema é perfeitamente real e suas soluções reais podem ser obtidas sem dificuldade.

Da primeira equação,

$$
x'=-y.
$$

Derivando,

$$
x''=-y'.
$$

Como

$$
y'=x,
$$

temos

$$
x''=-x.
$$

Portanto,

$$
x''+x=0.
$$

Assim,

$$
x(t)=C_1\cos t+C_2\sin t.
$$

Como

$$
y=-x',
$$

obtemos

$$
y(t)=C_1\sin t-C_2\cos t.
$$

Além disso,

$$
\frac{d}{dt}(x^2+y^2)
=
2xx'+2yy'
=
2x(-y)+2y(x)
=
0.
$$

Logo,

$$
\boxed{x^2+y^2=C.}
$$

As trajetórias são circunferências.

A origem é chamada de **centro**.

## 9. O sentido da rotação não vem apenas dos autovalores

Os autovalores

$$
\lambda=\pm i
$$

nos dizem que existe comportamento oscilatório sem crescimento nem decaimento exponencial.

Mas eles não determinam, sozinhos, se a rotação no plano ocorre no sentido horário ou anti-horário.

Para isso, basta examinar o campo em um ponto simples.

No ponto

$$
(1,0),
$$

temos

$$
(x',y')=(0,1).
$$

Portanto, o vetor aponta para cima e a rotação é **anti-horária**.

Esse pequeno teste será útil em sistemas com autovalores complexos.

# Parte V — Autovalores complexos com parte real não nula

## 10. Acrescentando expansão ou contração à rotação

Considere o sistema

$$
\begin{cases}
x'=\alpha x-\beta y,\\
y'=\beta x+\alpha y,
\end{cases}
$$

com $\beta\neq0$.

A matriz é

$$
A=
\begin{pmatrix}
\alpha&-\beta\\
\beta&\alpha
\end{pmatrix}.
$$

Seu polinômio característico é

$$
(\alpha-\lambda)^2+\beta^2=0.
$$

Logo,

$$
\boxed{
\lambda_{1,2}
=
\alpha\pm i\beta.
}
$$

Vamos descobrir o significado geométrico de $\alpha$ e $\beta$.

## 11. Resolvendo o sistema

Uma forma particularmente clara de escrever a solução é

$$
\begin{pmatrix}
x(t)\\
y(t)
\end{pmatrix}
=
e^{\alpha t}
\begin{pmatrix}
\cos(\beta t)&-\sin(\beta t)\\
\sin(\beta t)&\cos(\beta t)
\end{pmatrix}
\begin{pmatrix}
x_0\\
y_0
\end{pmatrix}.
$$

Observe os dois fatores.

O primeiro,

$$
e^{\alpha t},
$$

controla o tamanho da solução.

O segundo,

$$
\begin{pmatrix}
\cos(\beta t)&-\sin(\beta t)\\
\sin(\beta t)&\cos(\beta t)
\end{pmatrix},
$$

é uma matriz de rotação.

Portanto:

$$
\boxed{
\operatorname{Re}(\lambda)=\alpha
\quad\text{controla crescimento ou decaimento},
}
$$

enquanto

$$
\boxed{
|\operatorname{Im}(\lambda)|=|\beta|
\quad\text{controla a frequência angular da rotação}.
}
$$

## 12. Três possibilidades

### Caso 1 — $\alpha<0$

Então

$$
e^{\alpha t}\to0.
$$

As trajetórias giram enquanto se aproximam da origem.

Temos um

$$
\boxed{\text{foco estável}}
$$

ou **espiral estável**.

### Caso 2 — $\alpha>0$

Então

$$
e^{\alpha t}\to\infty.
$$

As trajetórias giram enquanto se afastam da origem.

Temos um

$$
\boxed{\text{foco instável}}
$$

ou **espiral instável**.

### Caso 3 — $\alpha=0$

Não há crescimento nem decaimento exponencial.

No caso canônico acima, as trajetórias são circunferências e temos um

$$
\boxed{\text{centro}.}
$$

```python
def retrato(alpha, beta=2.0, lim=3):
    x = np.linspace(-lim,lim,30)
    y = np.linspace(-lim,lim,30)
    X,Y = np.meshgrid(x,y)
    U = alpha*X - beta*Y
    V = beta*X + alpha*Y

    plt.figure(figsize=(7,7))
    plt.streamplot(X,Y,U,V,density=1.2)
    plt.axhline(0,linewidth=.8)
    plt.axvline(0,linewidth=.8)
    plt.xlim(-lim,lim)
    plt.ylim(-lim,lim)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(fr"$\alpha={alpha},\ \beta={beta}$")
    plt.grid(alpha=.2)
    plt.show()

retrato(-0.4)
retrato(0.0)
retrato(0.4)
```

# Parte VI — De uma matriz geral ao retrato de fase

## 13. Autovalores complexos de uma matriz real

Para uma matriz real

$$
A=
\begin{pmatrix}
a&b\\
c&d
\end{pmatrix},
$$

se os autovalores não são reais, eles aparecem em pares conjugados:

$$
\lambda_{1,2}
=
\alpha\pm i\beta,
\qquad
\beta\neq0.
$$

A solução real não precisa ser escrita com números complexos.

Se

$$
\mathbf v=\mathbf p+i\mathbf q
$$

é um autovetor associado a

$$
\lambda=\alpha+i\beta,
$$

então uma solução complexa é

$$
e^{(\alpha+i\beta)t}(\mathbf p+i\mathbf q).
$$

Usando

$$
e^{i\beta t}
=
\cos(\beta t)+i\sin(\beta t),
$$

e separando partes real e imaginária, obtemos duas soluções reais:

$$
\boxed{
\mathbf{x}_1(t)
=
e^{\alpha t}
\left[
\mathbf p\cos(\beta t)
-
\mathbf q\sin(\beta t)
\right]
}
$$

e

$$
\boxed{
\mathbf{x}_2(t)
=
e^{\alpha t}
\left[
\mathbf p\sin(\beta t)
+
\mathbf q\cos(\beta t)
\right].
}
$$

A solução real geral é

$$
\boxed{
\mathbf{x}(t)
=
c_1\mathbf{x}_1(t)
+
c_2\mathbf{x}_2(t).
}
$$

A geometria continua sendo determinada principalmente pelo sinal de $\alpha$.

## 14. Exemplo completo com autovalores complexos

Considere

$$
A=
\begin{pmatrix}
1&-2\\
2&1
\end{pmatrix}.
$$

O polinômio característico é

$$
\det(A-\lambda I)
=
(1-\lambda)^2+4.
$$

Portanto,

$$
\lambda_{1,2}
=
1\pm2i.
$$

A parte real é positiva:

$$
\operatorname{Re}(\lambda)=1>0.
$$

Assim, antes mesmo de escrever a solução completa, podemos prever:

$$
\boxed{\text{a origem é um foco instável}.}
$$

A solução é

$$
\begin{pmatrix}
x(t)\\y(t)
\end{pmatrix}
=
e^t
\begin{pmatrix}
\cos2t&-\sin2t\\
\sin2t&\cos2t
\end{pmatrix}
\begin{pmatrix}
x_0\\y_0
\end{pmatrix}.
$$

O fator $e^t$ produz expansão e o fator trigonométrico produz rotação.

No ponto $(1,0)$,

$$
A
\begin{pmatrix}
1\\0
\end{pmatrix}
=
\begin{pmatrix}
1\\2
\end{pmatrix},
$$

que aponta para cima.

Portanto, a rotação ocorre no sentido anti-horário.

# Parte VII — O papel da parte real dos autovalores

## 15. Uma primeira síntese espectral

Para os casos não degenerados estudados nesta aula:

### Dois autovalores reais negativos

$$
\lambda_1<0,\qquad\lambda_2<0
$$

produzem um **nó estável**.

### Dois autovalores reais positivos

$$
\lambda_1>0,\qquad\lambda_2>0
$$

produzem um **nó instável**.

### Autovalores reais de sinais opostos

$$
\lambda_1\lambda_2<0
$$

produzem uma **sela**.

### Autovalores complexos

$$
\lambda=\alpha\pm i\beta,
\qquad\beta\neq0.
$$

Então:

$$
\alpha<0
\Rightarrow
\text{foco estável},
$$

$$
\alpha>0
\Rightarrow
\text{foco instável},
$$

e

$$
\alpha=0
\Rightarrow
\text{centro}
$$

para sistemas lineares planares com um par puramente imaginário.

Uma mensagem começa a emergir:

$$
\boxed{
\text{o sinal das partes reais dos autovalores controla a estabilidade.}
}
$$

## 16. Estabilidade assintótica no caso linear

Para os casos considerados hoje, se todos os autovalores satisfazem

$$
\operatorname{Re}(\lambda)<0,
$$

então

$$
\mathbf{x}(t)\to0
\qquad
(t\to+\infty).
$$

A origem é **assintoticamente estável**.

Se pelo menos um autovalor possui

$$
\operatorname{Re}(\lambda)>0,
$$

existe uma direção na qual aparecem soluções que se afastam da origem.

A origem é **instável**.

O caso

$$
\operatorname{Re}(\lambda)=0
$$

é mais delicado.

O centro estudado hoje é estável, mas não assintoticamente estável:

as soluções próximas permanecem próximas, porém não convergem para a origem.

Na próxima aula veremos por que autovalores repetidos e casos de parte real zero exigem cuidado adicional.

# Parte VIII — Como analisar um sistema planar

## 17. Procedimento prático

Dado

$$
\mathbf{x}'=A\mathbf{x},
$$

com $A$ uma matriz real $2\times2$:

**Passo 1.** Encontre os autovalores de $A$.

**Passo 2.** Determine se são:

- reais e distintos;
- complexos conjugados;
- repetidos.

**Passo 3.** Se forem reais e distintos, encontre os autovetores.

Esses autovetores fornecem as direções invariantes do retrato de fase.

**Passo 4.** Examine os sinais dos autovalores reais ou das partes reais dos autovalores complexos.

**Passo 5.** Construa a solução geral.

**Passo 6.** Retorne à geometria:

- quais direções são invariantes?
- as trajetórias aproximam-se ou afastam-se da origem?
- há rotação?
- qual é o sentido da rotação?
- qual termo domina para tempos grandes?

# Parte XII — Autovalores repetidos

## 21. O caso que faltava

Até agora estudamos:

- dois autovalores reais distintos;
- um par de autovalores complexos conjugados.

Resta analisar o caso em que o polinômio característico possui uma raiz real dupla:

$$
\lambda_1=\lambda_2=\lambda.
$$

Aqui surge uma nova questão.

Saber que o autovalor é repetido **não é suficiente** para determinar o retrato de fase.

Precisamos perguntar:

> **Quantos autovetores linearmente independentes existem associados a $\lambda$?**

Em dimensão dois, há duas possibilidades:

1. existem dois autovetores linearmente independentes;
2. existe apenas uma direção própria.

Esses dois casos produzem geometrias diferentes.

## 22. Autovalor repetido com dois autovetores independentes

Suponha que

$$
A\mathbf v_1=\lambda\mathbf v_1,
\qquad
A\mathbf v_2=\lambda\mathbf v_2,
$$

com $\mathbf v_1$ e $\mathbf v_2$ linearmente independentes.

Como eles formam uma base de $\mathbb R^2$, todo vetor pode ser escrito como

$$
\mathbf{x}=c_1\mathbf v_1+c_2\mathbf v_2.
$$

Então

$$
A\mathbf{x}
=
c_1A\mathbf v_1+c_2A\mathbf v_2
=
\lambda(c_1\mathbf v_1+c_2\mathbf v_2).
$$

Logo,

$$
A\mathbf{x}=\lambda\mathbf{x}
$$

para todo $\mathbf{x}\in\mathbb R^2$.

Portanto,

$$
\boxed{A=\lambda I.}
$$

A solução é simplesmente

$$
\boxed{
\mathbf{x}(t)=e^{\lambda t}\mathbf{x}_0.
}
$$

Cada reta que passa pela origem é invariante.

Se $\lambda<0$, todas as trajetórias caminham radialmente para a origem.

Se $\lambda>0$, todas se afastam radialmente da origem.

Esse caso é frequentemente chamado de **nó estrela**.

```python
import numpy as np
import matplotlib.pyplot as plt

def campo_linear(A, titulo, lim=3):
    x = np.linspace(-lim, lim, 31)
    y = np.linspace(-lim, lim, 31)
    X, Y = np.meshgrid(x, y)
    U = A[0,0]*X + A[0,1]*Y
    V = A[1,0]*X + A[1,1]*Y

    plt.figure(figsize=(7,7))
    plt.streamplot(X, Y, U, V, density=1.2)
    plt.axhline(0, linewidth=.8)
    plt.axvline(0, linewidth=.8)
    plt.xlim(-lim,lim)
    plt.ylim(-lim,lim)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(titulo)
    plt.grid(alpha=.2)
    plt.show()

campo_linear(np.array([[-2.,0.],[0.,-2.]]),
             r"Nó estrela estável: $A=-2I$")
```

## 23. Autovalor repetido com apenas um autovetor

Considere agora

$$
A=
\begin{pmatrix}
\lambda&1\\
0&\lambda
\end{pmatrix}.
$$

O único autovalor é $\lambda$.

Para encontrar os autovetores,

$$
(A-\lambda I)\mathbf v=0,
$$

isto é,

$$
\begin{pmatrix}
0&1\\
0&0
\end{pmatrix}
\begin{pmatrix}
v_1\\v_2
\end{pmatrix}
=0.
$$

Logo,

$$
v_2=0,
$$

e existe apenas uma direção própria:

$$
\mathbf v=
\begin{pmatrix}
1\\0
\end{pmatrix}.
$$

A matriz não possui uma base de autovetores.

Dizemos que ela é **não diagonalizável** ou **defeituosa**.

## 24. Como resolver o caso defeituoso?

Escreva

$$
A=\lambda I+N,
$$

onde

$$
N=
\begin{pmatrix}
0&1\\
0&0
\end{pmatrix}.
$$

Observe que

$$
N^2=0.
$$

Como $\lambda I$ comuta com $N$,

$$
e^{At}
=
e^{(\lambda I+N)t}
=
e^{\lambda t}e^{Nt}.
$$

Mas

$$
e^{Nt}
=
I+tN+\frac{t^2N^2}{2!}+\cdots.
$$

Como $N^2=0$,

$$
e^{Nt}=I+tN.
$$

Portanto,

$$
\boxed{
e^{At}
=
e^{\lambda t}
\begin{pmatrix}
1&t\\
0&1
\end{pmatrix}.
}
$$

Para

$$
\mathbf{x}_0=
\begin{pmatrix}
x_0\\y_0
\end{pmatrix},
$$

temos

$$
\boxed{
\mathbf{x}(t)
=
e^{\lambda t}
\begin{pmatrix}
x_0+t y_0\\
y_0
\end{pmatrix}.
}
$$

Surge agora um fator novo:

$$
\boxed{t e^{\lambda t}.}
$$

Esse fator é a assinatura do caso não diagonalizável.

## 25. Interpretação geométrica do nó defeituoso

Se $\lambda<0$, então

$$
e^{\lambda t}\to0
$$

e também

$$
te^{\lambda t}\to0.
$$

Portanto, todas as soluções convergem para a origem.

Temos ainda um nó estável.

Entretanto, existe apenas **uma direção própria**.

As trajetórias que não começam exatamente sobre essa reta curvam-se e tornam-se assintoticamente paralelas a ela quando se aproximam da origem.

Se $\lambda>0$, o processo ocorre no sentido oposto e temos um nó instável.

Esse retrato é frequentemente chamado de **nó impróprio** ou **nó degenerado**, dependendo da terminologia adotada.

```python
A = np.array([[-1., 1.],
              [ 0.,-1.]])

campo_linear(A, r"Autovalor duplo $\lambda=-1$ com uma única direção própria")
```

## 26. Autovetor generalizado

Existe outra maneira importante de compreender o termo

$$
te^{\lambda t}.
$$

Se $\mathbf v$ é um autovetor,

$$
(A-\lambda I)\mathbf v=0,
$$

podemos procurar um vetor $\mathbf w$ tal que

$$
\boxed{
(A-\lambda I)\mathbf w=\mathbf v.
}
$$

Esse vetor é chamado de **autovetor generalizado**.

No caso de uma cadeia de comprimento dois,

$$
A\mathbf v=\lambda\mathbf v,
$$

$$
A\mathbf w=\lambda\mathbf w+\mathbf v.
$$

Então duas soluções independentes podem ser escritas como

$$
\mathbf{x}_1(t)=e^{\lambda t}\mathbf v
$$

e

$$
\boxed{
\mathbf{x}_2(t)
=
e^{\lambda t}
\left(
t\mathbf v+\mathbf w
\right).
}
$$

Assim, mesmo quando não existem dois autovetores independentes, ainda podemos construir duas soluções independentes utilizando um autovetor e um autovetor generalizado.

Essa é a primeira aparição, em nosso curso, da estrutura que em Álgebra Linear conduz à **forma de Jordan**.

# Parte XIII — Traço, determinante e discriminante

## 27. Podemos classificar sem calcular explicitamente os autovalores?

Considere

$$
A=
\begin{pmatrix}
a&b\\
c&d
\end{pmatrix}.
$$

O polinômio característico é

$$
p(\lambda)
=
\det(\lambda I-A).
$$

Calculando,

$$
p(\lambda)
=
\lambda^2-(a+d)\lambda+(ad-bc).
$$

Definimos

$$
\boxed{\tau=\operatorname{tr}(A)=a+d}
$$

e

$$
\boxed{\Delta=\det(A)=ad-bc.}
$$

Assim,

$$
\boxed{
p(\lambda)=\lambda^2-\tau\lambda+\Delta.
}
$$

Consequentemente,

$$
\boxed{
\lambda_{1,2}
=
\frac{\tau\pm\sqrt{\tau^2-4\Delta}}{2}.
}
$$

A natureza dos autovalores é controlada pelo discriminante

$$
\boxed{
D=\tau^2-4\Delta.
}
$$

## 28. O significado de $\tau$ e $\Delta$

Como

$$
\lambda_1+\lambda_2=\tau,
$$

o traço é a soma dos autovalores.

Além disso,

$$
\lambda_1\lambda_2=\Delta.
$$

Portanto, o determinante é o produto dos autovalores.

Essas duas relações já fornecem muita informação.

### Se $\Delta<0$

Então

$$
\lambda_1\lambda_2<0.
$$

Os autovalores reais possuem sinais opostos.

Logo,

$$
\boxed{\Delta<0\quad\Longrightarrow\quad\text{sela}.}
$$

### Se $\Delta>0$

Os autovalores reais, quando existirem, possuem o mesmo sinal.

Nesse caso, o traço ajuda a decidir qual:

$$
\tau<0
\Rightarrow
\text{parte real negativa},
$$

$$
\tau>0
\Rightarrow
\text{parte real positiva}.
$$

### Se $\Delta=0$

Pelo menos um autovalor é zero.

Nesse caso, a origem não é um equilíbrio isolado em um sistema linear homogêneo quando $A$ é singular: existe um núcleo não trivial,

$$
A\mathbf v=0,
$$

e todo ponto dessa direção é um equilíbrio.

## 29. O papel do discriminante

O sinal de

$$
D=\tau^2-4\Delta
$$

determina a natureza dos autovalores.

### $D>0$

Existem dois autovalores reais distintos.

### $D=0$

Existe um autovalor real duplo:

$$
\lambda=\frac{\tau}{2}.
$$

Será necessário verificar se existem um ou dois autovetores independentes.

### $D<0$

Os autovalores são complexos conjugados:

$$
\lambda_{1,2}
=
\frac{\tau}{2}
\pm
i\frac{\sqrt{4\Delta-\tau^2}}{2}.
$$

Portanto,

$$
\boxed{
\operatorname{Re}(\lambda)
=
\frac{\tau}{2}.
}
$$

Assim, para focos e centros, o sinal do traço determina diretamente crescimento ou decaimento.

# Parte XIV — O plano traço--determinante

## 30. Uma classificação global

Cada matriz real $2\times2$ determina um ponto

$$
(\tau,\Delta)
$$

no plano traço--determinante.

A curva

$$
\boxed{
\Delta=\frac{\tau^2}{4}
}
$$

corresponde a

$$
D=\tau^2-4\Delta=0.
$$

Ela separa a região de autovalores reais da região de autovalores complexos.

Temos:

- abaixo de $\Delta=0$: selas;
- entre $\Delta=0$ e $\Delta=\tau^2/4$: nós com autovalores reais distintos;
- acima de $\Delta=\tau^2/4$: focos ou centros;
- sobre $\Delta=\tau^2/4$: autovalor real repetido;
- sobre $\tau=0$ com $\Delta>0$: centros;
- sobre $\Delta=0$: pelo menos um autovalor zero.

Esse diagrama organiza praticamente toda a classificação linear planar.

```python
tau = np.linspace(-6,6,500)
parabola = tau**2/4

plt.figure(figsize=(9,7))
plt.plot(tau, parabola, linewidth=2, label=r"$\Delta=\tau^2/4$")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.text(-4.7,-2.0,"SELAS",fontsize=12)
plt.text(-4.8,2.0,"NÓS\nESTÁVEIS",fontsize=11)
plt.text(3.1,2.0,"NÓS\nINSTÁVEIS",fontsize=11)
plt.text(-3.4,7.0,"FOCOS\nESTÁVEIS",fontsize=11)
plt.text(2.0,7.0,"FOCOS\nINSTÁVEIS",fontsize=11)
plt.text(0.15,5.0,"centros sobre $\\tau=0$",fontsize=10)

plt.xlim(-6,6)
plt.ylim(-4,10)
plt.xlabel(r"Traço $\tau$")
plt.ylabel(r"Determinante $\Delta$")
plt.title("Plano traço–determinante")
plt.grid(alpha=.2)
plt.legend()
plt.show()
```

## 31. Lendo o plano traço--determinante

Vamos justificar cuidadosamente cada região.

### Região 1 — $\Delta<0$

O produto dos autovalores é negativo:

$$
\lambda_1\lambda_2<0.
$$

Logo, um é positivo e o outro negativo.

Temos uma

$$
\boxed{\text{sela}.}
$$

A sela é sempre instável.

---

### Região 2 — $\Delta>0$, $D>0$ e $\tau<0$

Os autovalores são reais, distintos e possuem o mesmo sinal.

Como sua soma é negativa, ambos são negativos.

Temos um

$$
\boxed{\text{nó estável}.}
$$

---

### Região 3 — $\Delta>0$, $D>0$ e $\tau>0$

Os autovalores são reais, distintos e ambos positivos.

Temos um

$$
\boxed{\text{nó instável}.}
$$

---

### Região 4 — $D<0$ e $\tau<0$

Os autovalores são

$$
\lambda=\frac{\tau}{2}\pm i\beta
$$

com parte real negativa.

Temos um

$$
\boxed{\text{foco estável}.}
$$

---

### Região 5 — $D<0$ e $\tau>0$

A parte real é positiva.

Temos um

$$
\boxed{\text{foco instável}.}
$$

---

### Região 6 — $\tau=0$, $\Delta>0$

Os autovalores são

$$
\lambda=\pm i\sqrt{\Delta}.
$$

No sistema linear planar temos um

$$
\boxed{\text{centro}.}
$$

Ele é estável, mas não assintoticamente estável.

## 32. E sobre a parábola $D=0$?

Sobre

$$
\Delta=\frac{\tau^2}{4},
$$

temos um autovalor duplo

$$
\lambda=\frac{\tau}{2}.
$$

Se

$$
\tau<0,
$$

o autovalor é negativo e o equilíbrio é um nó estável.

Se

$$
\tau>0,
$$

o autovalor é positivo e o equilíbrio é um nó instável.

Mas o ponto $(\tau,\Delta)$ **não distingue**:

- o nó estrela, com dois autovetores independentes;
- o nó defeituoso, com apenas uma direção própria.

Ambos possuem o mesmo traço e o mesmo determinante.

Portanto:

> **Traço e determinante classificam grande parte da dinâmica, mas não contêm toda a informação geométrica da matriz.**

No caso de autovalor repetido, ainda precisamos examinar os autovetores.

# Parte XV — Classificação completa dos equilíbrios hiperbólicos

## 33. O conceito de hiperbólico

Um equilíbrio linear é chamado de **hiperbólico** quando nenhum autovalor possui parte real zero:

$$
\boxed{
\operatorname{Re}(\lambda_j)\neq0
\quad
\text{para todo }j.
}
$$

No plano, são hiperbólicos:

- nós estáveis;
- nós instáveis;
- focos estáveis;
- focos instáveis;
- selas.

Não são hiperbólicos:

- centros;
- casos com autovalor zero.

Os casos hiperbólicos são particularmente importantes porque sua geometria é robusta sob pequenas perturbações.

Essa ideia ganhará ainda mais importância quando estudarmos sistemas não lineares.

## 34. Tabela de classificação

Para um sistema

$$
\mathbf{x}'=A\mathbf{x},
$$

com

$$
\tau=\operatorname{tr}(A),
\qquad
\Delta=\det(A),
\qquad
D=\tau^2-4\Delta,
$$

podemos resumir:

| Condição | Autovalores | Tipo |
|---|---|---|
| $\Delta<0$ | reais, sinais opostos | sela |
| $\Delta>0$, $D>0$, $\tau<0$ | reais negativos | nó estável |
| $\Delta>0$, $D>0$, $\tau>0$ | reais positivos | nó instável |
| $\Delta>0$, $D<0$, $\tau<0$ | complexos, parte real negativa | foco estável |
| $\Delta>0$, $D<0$, $\tau>0$ | complexos, parte real positiva | foco instável |
| $\Delta>0$, $D<0$, $\tau=0$ | puramente imaginários | centro |
| $D=0$, $\tau<0$ | real duplo negativo | nó estável; examinar autovetores |
| $D=0$, $\tau>0$ | real duplo positivo | nó instável; examinar autovetores |
| $\Delta=0$ | pelo menos um zero | caso não hiperbólico |

A tabela deve ser entendida como consequência das fórmulas para os autovalores, e não apenas memorizada.

# Parte XVI — Um exemplo usando apenas traço e determinante

## 35. Classificação antes de calcular os autovetores

Considere

$$
A=
\begin{pmatrix}
-2&-5\\
1&-2
\end{pmatrix}.
$$

Temos

$$
\tau=-4
$$

e

$$
\Delta=(-2)(-2)-(-5)(1)=9.
$$

Logo,

$$
D=\tau^2-4\Delta
=
16-36
=
-20<0.
$$

Portanto os autovalores são complexos.

Além disso,

$$
\tau<0.
$$

Assim, sem calcular explicitamente os autovalores, já concluímos:

$$
\boxed{\text{a origem é um foco estável}.}
$$

Se quisermos os autovalores,

$$
\lambda
=
\frac{-4\pm\sqrt{-20}}{2}
=
-2\pm i\sqrt5.
$$

O cálculo explícito confirma a classificação.

# Parte XVII — Cuidado com a fronteira da classificação

## 36. Parte real zero é um caso especial

Para sistemas lineares, um par

$$
\lambda=\pm i\beta
$$

produz um centro.

Mas é importante guardar uma advertência para o futuro.

Quando estudarmos um sistema **não linear**

$$
\mathbf{x}'=\mathbf F(\mathbf{x}),
$$

a linearização em um equilíbrio pode ter autovalores puramente imaginários.

Nesse caso, a parte linear sozinha geralmente **não permite concluir** que o sistema não linear possui um centro.

De maneira semelhante, um autovalor zero torna a linearização inconclusiva para várias questões de estabilidade.

Portanto, a regra

$$
\text{“olhe os autovalores da linearização”}
$$

é particularmente poderosa quando o equilíbrio é hiperbólico.

Essa distinção será fundamental quando passarmos de sistemas lineares para sistemas não lineares.

## 37. Classificador pelo traço e determinante

O código abaixo calcula

\tau,\qquad \Delta,\qquad D

e fornece a classificação preliminar.

Nos casos de autovalor repetido, ele nos lembra que ainda precisamos verificar a dimensão do autoespaço.

```python
import numpy as np

def classificar_2x2(A, tol=1e-10):
    A = np.asarray(A, dtype=float)

    tau = np.trace(A)
    Delta = np.linalg.det(A)
    D = tau**2 - 4*Delta
    vals = np.linalg.eigvals(A)

    print("A =")
    print(A)
    print(f"\ntraço τ = {tau:.6g}")
    print(f"determinante Δ = {Delta:.6g}")
    print(f"discriminante D = {D:.6g}")
    print("autovalores =", vals)

    if Delta < -tol:
        tipo = "sela"
    elif abs(Delta) <= tol:
        tipo = "caso não hiperbólico: pelo menos um autovalor zero"
    elif D > tol:
        if tau < 0:
            tipo = "nó estável"
        elif tau > 0:
            tipo = "nó instável"
        else:
            tipo = "caso degenerado"
    elif D < -tol:
        if tau < -tol:
            tipo = "foco estável"
        elif tau > tol:
            tipo = "foco instável"
        else:
            tipo = "centro"
    else:
        if tau < 0:
            tipo = "autovalor duplo negativo: nó estável; verificar autovetores"
        elif tau > 0:
            tipo = "autovalor duplo positivo: nó instável; verificar autovetores"
        else:
            tipo = "caso degenerado"

    print("\nClassificação:", tipo)

A = np.array([[-2., -5.],
              [ 1., -2.]])

classificar_2x2(A)
```

# 39. Síntese completa da aula

Nesta aula construímos a classificação dos sistemas lineares planares

$$
\mathbf{x}'=A\mathbf{x}.
$$

A análise começou pela geometria dos autovalores e autovetores:

$$
\boxed{
\text{autovalores}
\longrightarrow
\text{crescimento, decaimento e oscilação}
}
$$

e

$$
\boxed{
\text{autovetores}
\longrightarrow
\text{direções invariantes}.
}
$$

Para autovalores reais distintos:

$$
\lambda_1,\lambda_2<0
\Rightarrow
\text{nó estável},
$$

$$
\lambda_1,\lambda_2>0
\Rightarrow
\text{nó instável},
$$

$$
\lambda_1\lambda_2<0
\Rightarrow
\text{sela}.
$$

Para

$$
\lambda=\alpha\pm i\beta,
$$

temos

$$
\alpha<0
\Rightarrow
\text{foco estável},
$$

$$
\alpha=0
\Rightarrow
\text{centro},
$$

$$
\alpha>0
\Rightarrow
\text{foco instável}.
$$

Para um autovalor real duplo, precisamos distinguir:

- dois autovetores independentes: nó estrela;
- apenas um autovetor: nó defeituoso, com termos $te^{\lambda t}$.

Finalmente, introduzimos

$$
\tau=\operatorname{tr}(A),
\qquad
\Delta=\det(A),
$$

e

$$
D=\tau^2-4\Delta,
$$

obtendo uma classificação organizada no **plano traço--determinante**.

A mensagem conceitual final é:

$$
\boxed{
A
\longrightarrow
(\lambda,\mathbf v)
\longrightarrow
e^{At}
\longrightarrow
\text{retrato de fase}
\longrightarrow
\text{estabilidade}.
}
$$

A partir daqui temos a base necessária para avançar para sistemas não lineares e compreender por que a linearização em torno de um equilíbrio pode revelar sua dinâmica local.
