# 3. Equações com Variáveis Separáveis

Nesta seção estudaremos a primeira classe importante de equações diferenciais de primeira ordem para a qual dispomos de um procedimento sistemático de integração.

A ideia básica é simples: transformar a equação de modo que todos os termos envolvendo $y$ fiquem de um lado e todos os termos envolvendo $x$ fiquem do outro.

Entretanto, há um ponto matemático importante que merece atenção:

> ao dividirmos uma equação por uma expressão que pode se anular, podemos perder soluções.

Por isso, além do procedimento algébrico, discutiremos o domínio das soluções e as chamadas **soluções constantes** ou **soluções de equilíbrio**.

## 3.1 Variáveis separadas e variáveis separáveis

Uma equação da forma

$$
P(y)\,dy = Q(x)\,dx
$$

é dita uma equação com **variáveis separadas**.

Mais frequentemente encontraremos uma equação escrita na forma

$$
y'=f(x)g(y),
$$

isto é,

$$
\frac{dy}{dx}=f(x)g(y).
$$

Em regiões onde $g(y)\neq 0$, podemos escrever formalmente

$$
\frac{1}{g(y)}\,dy=f(x)\,dx.
$$

As variáveis ficaram separadas: de um lado aparece apenas $y$, e do outro apenas $x$.

Integrando,

$$
\int \frac{1}{g(y)}\,dy
=
\int f(x)\,dx + C.
$$

Em geral, essa relação define a solução **implicitamente**.

O livro de Kiselev apresenta a classe em uma forma ainda mais geral, na qual os coeficientes de $dx$ e $dy$ se fatoram em funções de $x$ e de $y$. O princípio é o mesmo: depois de uma divisão apropriada, obtém-se uma equação com variáveis separadas.

## 3.2 Um cuidado

Considere

$$
y'=f(x)g(y).
$$

O passo

$$
\frac{dy}{g(y)}=f(x)\,dx
$$

pressupõe que

$$
g(y)\neq 0.
$$

Portanto, antes de dividir por $g(y)$, devemos verificar os zeros de $g$.

Se

$$
g(y_*)=0,
$$

então a função constante

$$
\boxed{y(x)=y_*}
$$

satisfaz

$$
y'=0=f(x)g(y_*).
$$

Assim, cada zero de $g$ fornece uma possível solução constante.

Essas soluções podem desaparecer do cálculo se dividirmos imediatamente por $g(y)$.

Esse é um dos cuidados mais importantes no uso do método de separação de variáveis.

### Exemplo

Considere

$$
y'=x(y-1).
$$

Temos

$$
g(y)=y-1.
$$

Como

$$
g(1)=0,
$$

a função

$$
\boxed{y(x)=1}
$$

é uma solução.

Se dividirmos a equação por $y-1$, essa solução deixa de aparecer no cálculo. Portanto, devemos registrá-la **antes** da divisão.

## 3.3 O procedimento básico

Para uma equação

$$
y'=f(x)g(y),
$$

podemos seguir o seguinte roteiro.

### Passo 1 — procure soluções constantes

Resolva

$$
g(y)=0.
$$

Cada raiz $y_*$ fornece uma candidata

$$
y(x)=y_* \, .
$$

### Passo 2 — em uma região onde $g(y) \neq 0$, separe as variáveis

$$
\frac{dy}{g(y)}=f(x)\,dx.
$$

### Passo 3 — integre

$$
\int\frac{dy}{g(y)}
=
\int f(x)\,dx + C \, .
$$

### Passo 4 — quando possível, isole $y$

A relação obtida pode fornecer uma solução explícita

$$
y=\phi(x,C),
$$

mas isso nem sempre é necessário ou conveniente.

### Passo 5 — determine o intervalo de validade

Uma solução (formal) de uma EDO deve ser considerada em um intervalo no qual:

- a solução encontrada (sua fórmula) esteja definida;
- a equação diferencial esteja definida;
- as operações utilizadas na dedução sejam válidas.

## 3.4 Exemplo

Resolva

$$
y'= \, x \, y.
$$

Antes de dividir por $y$, observamos que

$$
y=0
$$

é uma solução constante.

Agora suponha $y \neq 0$. Então

$$
\frac{dy}{y}=x\,dx.
$$

Integrando,

$$
\ln|y|=\frac{x^2}{2}+C.
$$

Exponentiando,

$$
|y|=e^C e^{x^2/2}.
$$

Absorvendo o sinal e o fator $e^C$ em uma nova constante real $A$,

$$
\boxed{y=Ae^{x^2/2}}.
$$

Observe que $A=0$ recupera a solução constante $y=0$.

Portanto, neste exemplo, a família final já inclui a solução que inicialmente poderia ter sido perdida pela divisão.

## 3.5 Visualizando a família de soluções

O computador será usado apenas para visualizar a família

$$
y=Ae^{x^2/2}.
$$

oberve que não estamos pedindo ao Python que resolva a equação diferencial.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 500)

plt.figure(figsize=(8, 6))

for A in [-2, -1, -0.5, 0, 0.5, 1, 2]:
    y = A*np.exp(x**2/2)
    plt.plot(x, y, label=fr"$A={A}$")

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Família de soluções de $y'=xy$")
plt.ylim(-12, 12)
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

Observe:

- se $A>0$, a solução permanece positiva;
- se $A<0$, permanece negativa;
- se $A=0$, temos a solução de equilíbrio;
- as curvas não atravessam a solução $y=0$.

Esse último fato também é compatível com a unicidade do problema de valor inicial, pois $f(x,y)=xy$ e $f_y(x,y)=x$ são contínuas em todo o plano.

## 3.6 Problema de valor inicial

Considere

$$
y'=x \, y,
\qquad
y(0)=2.
$$

A solução geral é

$$
y=A \, e^{x^2/2}.
$$

A condição inicial fornece

$$
2=A \, e^0,
$$

logo

$$
A=2.
$$

Portanto,

$$
\boxed{
y(x)=2e^{x^2/2}.
}
$$

A solução está definida para todo $x\in\mathbb R$.

## 3.7 Uma formulação mais rigorosa para o problema de valor inicial

Considere

$$
y'=f(x)g(y),
\qquad
y(x_0)=y_0,
$$

e suponha que $g(y_0)\neq0$.

Em um intervalo onde a solução não encontra um zero de $g$, podemos escrever

$$
\frac{y'(x)}{g(y(x))}=f(x).
$$

Escolha uma primitiva $G$ de $1/g$, isto é,

$$
G'(y)=\frac{1}{g(y)}.
$$

Pela regra da cadeia,

$$
\frac{d}{dx}G(y(x))
=
G'(y(x))y'(x)
=
\frac{y'(x)}{g(y(x))}
=
f(x).
$$

Integrando de $x_0$ até $x$,

$$
G(y(x))-G(y_0)
=
\int_{x_0}^{x}f(s)\,ds.
$$

Logo,

$$
\boxed{
G(y(x))
=
G(y_0)+\int_{x_0}^{x}f(s)\,ds.
}
$$

Essa forma tem uma vantagem conceitual: a constante de integração já está incorporada pela condição inicial.

Ela também deixa claro que a separação de variáveis pode ser justificada diretamente pela regra da cadeia, sem interpretar $dy$ e $dx$ como objetos algébricos independentes.

## 3.8 Segundo exemplo: uma equação com solução de equilíbrio

Considere

$$
y'=x(1-y^2).
$$

Primeiro procuramos os zeros de

$$
g(y)=1-y^2.
$$

Temos

$$
1-y^2=0
$$

quando

$$
y=\pm1.
$$

Portanto,

$$
\boxed{y=1}
\qquad\text{e}\qquad
\boxed{y=-1}
$$

são soluções constantes.

Agora, para $y\neq\pm1$,

$$
\frac{dy}{1-y^2}=x\,dx.
$$

Como

$$
\frac{1}{1-y^2}
=
\frac12\left(
\frac{1}{1+y}
+
\frac{1}{1-y}
\right),
$$

obtemos

$$
\int\frac{dy}{1-y^2}
=
\frac12\ln\left|\frac{1+y}{1-y}\right|.
$$

Assim,

$$
\frac12\ln\left|\frac{1+y}{1-y}\right|
=
\frac{x^2}{2}+C.
$$

Ou, equivalentemente,

$$
\boxed{
\ln\left|\frac{1+y}{1-y}\right|
=
x^2+C.
}
$$

Essa forma implícita já é uma solução perfeitamente válida.

## 3.9 O que a geometria nos diz antes da integração?

Na equação

$$
y'=x(1-y^2),
$$

as retas

$$
y=1
\qquad\text{e}\qquad
y=-1
$$

são soluções constantes.

Além disso, o sinal de $y'$ depende simultaneamente do sinal de $x$ e de $1-y^2$.

Para $|y|<1$,

$$
1-y^2>0,
$$

enquanto para $|y|>1$,

$$
1-y^2<0.
$$

Podemos visualizar o campo de direções e as duas soluções de equilíbrio.

```python
def campo_direcoes(f, xlim=(-3,3), ylim=(-3,3), nx=23, ny=23):
    xs = np.linspace(xlim[0], xlim[1], nx)
    ys = np.linspace(ylim[0], ylim[1], ny)
    X, Y = np.meshgrid(xs, ys)
    M = f(X, Y)

    U = 1/np.sqrt(1 + M**2)
    V = M/np.sqrt(1 + M**2)

    plt.figure(figsize=(8, 6))
    plt.quiver(X, Y, U, V, angles="xy", pivot="mid",
               headwidth=0, headlength=0, headaxislength=0)

    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(alpha=0.2)

campo_direcoes(lambda x, y: x*(1-y**2), (-3,3), (-2.5,2.5))

x = np.linspace(-3, 3, 400)
plt.plot(x, np.ones_like(x), linewidth=2.2, label=r"$y=1$")
plt.plot(x, -np.ones_like(x), linewidth=2.2, label=r"$y=-1$")

plt.title(r"Campo de direções de $y'=x(1-y^2)$")
plt.legend()
plt.show()
```

A visualização sugere que as soluções de equilíbrio funcionam como curvas que não podem ser atravessadas por outras soluções, desde que valha unicidade.

De fato,

$$
F(x,y)=x(1-y^2)
$$

e

$$
F_y(x,y)=-2xy
$$

são contínuas em todo o plano. Portanto, por cada ponto passa localmente uma única solução.

Assim, uma solução que começa com $-1<y_0<1$ não pode simplesmente atravessar $y=1$ ou $y=-1$.

## 3.10 Um exemplo inspirado no texto: $(1+e^x)yy'=e^x$

Considere

$$
(1+e^x)yy'=e^x.
$$

Como $1+e^x>0$ para todo $x$, podemos escrever

$$
yy'=\frac{e^x}{1+e^x}.
$$

Ou,

$$
y\,dy
=
\frac{e^x}{1+e^x}\,dx.
$$

Integrando,

$$
\frac{y^2}{2}
=
\ln(1+e^x)+C.
$$

Se impusermos

$$
y(0)=1,
$$

então

$$
\frac12=\ln 2+C.
$$

Logo,

$$
C=\frac12-\ln2.
$$

Portanto,

$$
\boxed{
\frac{y^2}{2}
=
\ln(1+e^x)+\frac12-\ln2.
}
$$

Podemos escrever

$$
y^2
=
1+2\ln\left(\frac{1+e^x}{2}\right).
$$

Como a condição inicial exige $y(0)=1>0$, escolhemos localmente o ramo positivo:

$$
\boxed{
y(x)=
\sqrt{
1+2\ln\left(\frac{1+e^x}{2}\right)
}.
}
$$

Mas agora surge uma questão que não deve ser ignorada: **para quais valores de $x$ essa expressão está definida?**

## 3.11 Fórmula não é sinônimo de solução global

A expressão

$$
y(x)=
\sqrt{
1+2\ln\left(\frac{1+e^x}{2}\right)
}
$$

só define uma função real quando

$$
1+2\ln\left(\frac{1+e^x}{2}\right)>0.
$$

Além disso, na equação original,

$$
(1+e^x)yy'=e^x,
$$

o lado direito nunca é zero. Portanto, uma solução diferenciável não pode atingir $y=0$, pois isso tornaria o lado esquerdo igual a zero.

Consequentemente, o intervalo de existência da solução que parte de $(0,1)$ termina quando a expressão sob a raiz tende a zero.

Esse exemplo ilustra uma distinção importante:

> obter uma expressão algébrica é apenas parte do problema; é preciso determinar em qual intervalo ela realmente representa uma solução da EDO.

```python
x = np.linspace(-4, 3, 800)

rad = 1 + 2*np.log((1 + np.exp(x))/2)

mask = rad > 0

plt.figure(figsize=(8, 5))
plt.plot(x[mask], np.sqrt(rad[mask]))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Solução do PVI $(1+e^x)yy'=e^x,\; y(0)=1$")
plt.grid(alpha=0.25)
plt.show()
```

O gráfico não determina o intervalo por nós; ele apenas ajuda a visualizar a aproximação da solução ao limite de seu domínio.

A determinação exata do extremo do intervalo deve ser feita analiticamente resolvendo

$$
1+2\ln\left(\frac{1+e^x}{2}\right)=0.
$$

## 3.12 Soluções implícitas podem ser naturais

Considere

$$
y'=\frac{x}{1+y^2}.
$$

Separando,

$$
(1+y^2)\,dy=x\,dx.
$$

Integrando,

$$
y+\frac{y^3}{3}
=
\frac{x^2}{2}+C.
$$

Não há necessidade de resolver explicitamente essa equação cúbica para $y$.

A relação

$$
\boxed{
y+\frac{y^3}{3}-\frac{x^2}{2}=C
}
$$

é uma descrição perfeitamente adequada da família de curvas integrais.

Além disso, como

$$
\frac{d}{dy}\left(y+\frac{y^3}{3}\right)
=
1+y^2>0,
$$

a função

$$
G(y)=y+\frac{y^3}{3}
$$

é estritamente crescente em $\mathbb R$.

Portanto, para cada valor do lado direito existe um único valor de $y$. A solução implícita determina, de fato, uma função $y(x)$ globalmente.

## 3.13 Visualizando uma família definida implicitamente

Podemos desenhar as curvas

$$
y+\frac{y^3}{3}-\frac{x^2}{2}=C
$$

sem resolver a equação cúbica para $y$.

```python
x = np.linspace(-3, 3, 500)
y = np.linspace(-3, 3, 500)
X, Y = np.meshgrid(x, y)

F = Y + Y**3/3 - X**2/2

plt.figure(figsize=(8, 6))
levels = [-3, -2, -1, 0, 1, 2, 3]
cs = plt.contour(X, Y, F, levels=levels)
plt.clabel(cs, inline=True, fontsize=8)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Curvas integrais implícitas de $y'=x/(1+y^2)$")
plt.grid(alpha=0.2)
plt.show()
```

Esse tipo de visualização é particularmente útil porque reforça uma ideia importante:

> integrar uma EDO não significa necessariamente obter $y$ isolado.

Uma relação implícita pode conter toda a informação necessária sobre a solução.

## 3.14 Separação com condição inicial: forma por integrais definidas

Para

$$
y'=f(x)g(y),
\qquad y(x_0)=y_0,
$$

quando $g(y)\neq0$ no intervalo considerado, podemos escrever diretamente

$$
\int_{y_0}^{y(x)}
\frac{du}{g(u)}
=
\int_{x_0}^{x} f(s)\,ds.
$$

Essa forma é particularmente elegante para problemas de valor inicial.

Ela evita introduzir uma constante arbitrária e deixa explícito que a integral do lado esquerdo depende do valor desconhecido $y(x)$.

Em muitos problemas teóricos, essa é a forma mais natural de escrever a solução.

## 3.15 Exemplo de crescimento exponencial

Considere

$$
y'=ky,
$$

onde $k$ é constante.

A solução $y=0$ é uma solução de equilíbrio.

Para $y\neq0$,

$$
\frac{dy}{y}=k\,dx.
$$

Integrando,

$$
\ln|y|=kx+C,
$$

de onde

$$
\boxed{y=Ae^{kx}}.
$$

Para o problema de valor inicial

$$
y(0)=y_0,
$$

temos

$$
\boxed{y(x)=y_0e^{kx}}.
$$

O sinal de $k$ determina qualitativamente o comportamento:

- $k>0$: crescimento exponencial;
- $k<0$: decaimento exponencial;
- $k=0$: soluções constantes.

```python
x = np.linspace(0, 5, 500)

plt.figure(figsize=(8, 5))

for k in [-1, -0.5, 0, 0.5, 1]:
    y = np.exp(k*x)
    plt.plot(x, y, label=fr"$k={k}$")

plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Soluções de $y'=ky$ com $y(0)=1$")
plt.ylim(0, 12)
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

## 3.16 Modelo logístico como equação separável

Uma aplicação clássica é

$$
P'=rP\left(1-\frac{P}{K}\right),
$$

com $r>0$ e $K>0$.

Antes de separar as variáveis, observamos que

$$
P=0
\qquad\text{e}\qquad
P=K
$$

são soluções de equilíbrio.

Para $P\neq0,K$,

$$
\frac{dP}{P(1-P/K)}=r\,dt.
$$

Como

$$
\frac{1}{P(1-P/K)}
=
\frac{1}{P}
+
\frac{1}{K-P},
$$

obtemos, após integração,

$$
\ln\left|\frac{P}{K-P}\right|
=
rt+C.
$$

O objetivo aqui não é aprofundar o modelo logístico, mas perceber que a separação de variáveis aparece naturalmente em problemas de crescimento.

```python
t = np.linspace(0, 10, 500)
K = 10
r = 0.8

plt.figure(figsize=(8, 5))

for P0 in [0.5, 2, 5, 8, 12]:
    A = (K-P0)/P0
    P = K/(1 + A*np.exp(-r*t))
    plt.plot(t, P, label=fr"$P_0={P0}$")

plt.axhline(K, linestyle="--", linewidth=1.5, label=r"$P=K$")
plt.xlabel("t")
plt.ylabel("P(t)")
plt.title(r"Algumas soluções da equação logística")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

A figura permite comparar soluções que começam abaixo e acima da capacidade de suporte $K$.

Mais uma vez, o gráfico apenas visualiza soluções cuja forma já foi obtida analiticamente.

## 3.17 Um erro frequente

Considere

$$
y'=y^2.
$$

Uma separação apressada produz

$$
\frac{dy}{y^2}=dx,
$$

e então

$$
-\frac1y=x+C,
$$

de onde

$$
y=-\frac{1}{x+C}.
$$

Essa família está correta para soluções não nulas.

Mas a divisão por $y^2$ eliminou a solução

$$
\boxed{y=0}.
$$

Assim, o conjunto das soluções inclui:

$$
\boxed{
y(x)=0
}
$$

e

$$
\boxed{
y(x)=-\frac{1}{x+C},
}
$$

em intervalos que não contenham o ponto $x=-C$.

Observe também que as soluções não nulas não são definidas em toda a reta real: elas apresentam uma singularidade em tempo finito.

```python
x_left = np.linspace(-4, -0.08, 500)
x_right = np.linspace(0.08, 4, 500)

plt.figure(figsize=(8, 6))
plt.plot(x_left, -1/x_left)
plt.plot(x_right, -1/x_right)
plt.axhline(0, linewidth=2, label=r"solução $y=0$")
plt.axvline(0, linestyle="--", linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Uma solução de $y'=y^2$ e sua singularidade")
plt.ylim(-8, 8)
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

## 3.18 Separabilidade e reconhecimento da estrutura

Nem toda equação aparece imediatamente na forma

$$
y'=f(x)g(y).
$$

Por exemplo,

$$
(1+x^2)y'=x(1+y^2)
$$

pode ser reescrita como

$$
\frac{dy}{dx}
=
\frac{x}{1+x^2}(1+y^2).
$$

Logo,

$$
\frac{dy}{1+y^2}
=
\frac{x}{1+x^2}\,dx.
$$

A habilidade central, portanto, não é memorizar uma fórmula, mas **reconhecer quando a dependência em $x$ e em $y$ pode ser fatorada**.

## 3.19 Resumo conceitual

Para equações da forma

$$
y'=f(x)g(y),
$$

o método de separação de variáveis consiste essencialmente em usar uma primitiva de $1/g$.

Se

$$
G'(y)=\frac1{g(y)},
$$

então, em regiões onde $g(y)\neq0$,

$$
\frac{d}{dx}G(y(x))=f(x).
$$

Portanto,

$$
G(y(x))=F(x)+C,
$$

onde $F'=f$.

Essa é a estrutura matemática por trás da manipulação simbólica

$$
\frac{dy}{g(y)}=f(x)\,dx.
$$

Os pontos essenciais são:

- identificar a separabilidade;
- verificar zeros dos fatores antes de dividir;
- integrar;
- aceitar soluções implícitas quando conveniente;
- impor condições iniciais;
- determinar corretamente o intervalo de definição da solução.

A separação de variáveis é frequentemente apresentada como uma regra :

$$
y'=f(x)g(y)
\quad\Longrightarrow\quad
\frac{dy}{g(y)}=f(x)\,dx.
$$

Por trás dessa notação há uma ideia mais rigorosa: encontrar uma função $G$ tal que

$$
G'(y)=\frac1{g(y)} \, ,
$$

isto é uma primitiva de $\frac1{g(y)}$,
e aplicar a regra da cadeia à composição $G(y(x))$.

Essa interpretação será útil quando começarmos a fazer mudanças de variável em classes mais gerais de equações.
