# 7. Bifurcações em Sistemas Unidimensionais

Na aula anterior estudamos equações autônomas $\dot{x}=f(x)$, seus pontos de equilíbrio e sua estabilidade. Agora permitiremos que o campo vetorial dependa de um parâmetro:

$$
\boxed{\dot{x}=f(x,r).}
$$

A pergunta central : **como a dinâmica muda quando variamos $r$?**

Pontos de equilíbrio podem surgir, desaparecer ou trocar de estabilidade. Uma mudança qualitativa desse tipo é chamada de **bifurcação**.

Seguiremos o Capítulo 3 de Steven H. Strogatz, concentrando-nos nas bifurcações elementares de pontos de equilíbrio em dimensão um.

Uma ferramenta útil para o que se segue é o conceito de **Forma Normal**

Forma normal = o modelo local mais simples que conserva a estrutura qualitativa da bifurcação.

## 7.1 Parâmetros e valores de bifurcação

Para cada valor fixado de $r$, a equação $\dot{x}=f(x,r)$ define uma EDO autônoma. Seus equilíbrios satisfazem

$$
\boxed{f(x^*,r)=0.}
$$

À medida que $r$ varia, os equilíbrios podem mudar de posição, aparecer, desaparecer ou mudar de estabilidade.

Chamamos $r=r_c$ de **valor de bifurcação** quando a dinâmica sofre uma mudança qualitativa.

# 7.2 Bifurcação sela-nó

A bifurcação **sela-nó** (*saddle-node*) é o mecanismo mais simples de criação ou destruição de pontos de equilíbrio. Sua forma normal pode ser escrita como

$$
\boxed{\dot{x}=r+x^2.}
$$

Para $r<0$,

$$
x_\pm^*=\pm\sqrt{-r}.
$$

Como $f_x=2x$, o equilíbrio $-\sqrt{-r}$ é estável e $+\sqrt{-r}$ é instável.

Para $r=0$,

$$
\dot{x}=x^2,
$$

e $x^*=0$ é semiestável.

Para $r>0$, $r+x^2>0$ para todo $x$ e não existem equilíbrios.

Portanto,

$$
\boxed{
\begin{array}{c|c}
r<0 & \text{dois equilíbrios}\\
r=0 & \text{um equilíbrio semiestável}\\
r>0 & \text{nenhum equilíbrio}
\end{array}}
$$

Os dois equilíbrios aproximam-se, colidem e desaparecem em $r_c=0$.

```python
import numpy as np
import matplotlib.pyplot as plt

def grafico_sela_no(r=-1.0):
    x=np.linspace(-2.5,2.5,700)
    plt.figure(figsize=(8,5))
    plt.plot(x,r+x**2)
    plt.axhline(0,linewidth=.9)
    plt.axvline(0,linewidth=.7)
    plt.xlabel("x"); plt.ylabel(r"$f(x,r)$")
    plt.title(fr"$\dot x=r+x^2$,  $r={r:.2f}$")
    plt.ylim(-2,5); plt.grid(alpha=.25)
    plt.show()

grafico_sela_no(-1)
```

## 7.3 Diagrama de bifurcação

Um **diagrama de bifurcação** representa os equilíbrios $x^*$ em função do parâmetro $r$.

Para a sela-nó,

$$
x^*=\pm\sqrt{-r},\qquad r\leq0.
$$

Usaremos linha contínua para ramos estáveis e linha tracejada para ramos instáveis.

```python
r=np.linspace(-3,0,500)
plt.figure(figsize=(8,6))
plt.plot(r,-np.sqrt(-r),linewidth=2,label="estável")
plt.plot(r,np.sqrt(-r),"--",linewidth=2,label="instável")
plt.axhline(0,linewidth=.7); plt.axvline(0,linewidth=.7)
plt.xlabel("r"); plt.ylabel(r"$x^*$")
plt.title(r"Diagrama de bifurcação de $\dot{x}=r+x^2$")
plt.grid(alpha=.25); plt.legend(); plt.show()
```

# 7.4 Por que $r+x^2$ é uma forma normal?

Considere $\dot{x}=f(x,r)$ e suponha que dois equilíbrios colidam em $(x_c,r_c)$. No ponto de colisão,

$$
f(x_c,r_c)=0
$$

e, pela tangência,

$$
f_x(x_c,r_c)=0.
$$

A expansão de Taylor fornece

$$
f(x,r)=f(x_c,r_c)+f_x(x_c,r_c)(x-x_c)
+f_r(x_c,r_c)(r-r_c)
+\frac12f_{xx}(x_c,r_c)(x-x_c)^2+\cdots.
$$

Logo,

$$
\dot{x}\approx a(r-r_c)+b(x-x_c)^2,
$$

onde

$$
a=f_r(x_c,r_c),\qquad b=\frac12f_{xx}(x_c,r_c).
$$

Quando $a\neq0$ e $b\neq0$, translações e mudanças de escala levam à forma normal

$$
\boxed{\dot{x}=r\pm x^2.}
$$

Assim, uma sela-nó genérica satisfaz

$$
f=0,\qquad f_x=0,\qquad f_r\neq0,\qquad f_{xx}\neq0
$$

no ponto crítico.

# 7.5 Bifurcação transcrítica

Em algumas aplicações existe um equilíbrio para todos os valores do parâmetro. Ele pode mudar de estabilidade sem desaparecer.

A forma normal é

$$
\boxed{\dot{x}=rx-x^2=x(r-x).}
$$

Os equilíbrios são

$$
\boxed{x_1^*=0,\qquad x_2^*=r.}
$$

Como

$$
f_x=r-2x,
$$

no ramo $x^*=0$ temos $f_x=r$: ele é estável para $r<0$ e instável para $r>0$.

No ramo $x^*=r$ temos $f_x=-r$: ele é instável para $r<0$ e estável para $r>0$.

Portanto, os dois ramos **trocam de estabilidade** em $r=0$.

```python
rn=np.linspace(-3,0,300); rp=np.linspace(0,3,300)
plt.figure(figsize=(8,6))
plt.plot(rn,0*rn,linewidth=2)
plt.plot(rp,0*rp,"--",linewidth=2)
plt.plot(rn,rn,"--",linewidth=2)
plt.plot(rp,rp,linewidth=2)
plt.axhline(0,linewidth=.7); plt.axvline(0,linewidth=.7)
plt.xlabel("r"); plt.ylabel(r"$x^*$")
plt.title(r"Transcrítica: $\dot{x}=rx-x^2$")
plt.grid(alpha=.25); plt.show()
```

## 7.6 Sela-nó versus transcrítica

Na sela-nó,

$$
\dot{x}=r+x^2,
$$

dois equilíbrios são criados ou destruídos juntos.

Na transcrítica,

$$
\dot{x}=rx-x^2,
$$

dois ramos persistem e **trocam estabilidade**.

Por isso, ao construir um diagrama de bifurcação, não basta localizar os equilíbrios: precisamos também classificá-los.

# 7.7 Pitchfork supercrítica

Considere

$$
\boxed{\dot{x}=rx-x^3=x(r-x^2).}
$$

Existe sempre o equilíbrio $x^*=0$. Para $r>0$ aparecem ainda

$$
\boxed{x_\pm^*=\pm\sqrt r.}
$$

Como

$$
f_x=r-3x^2,
$$

o equilíbrio $x=0$ é estável para $r<0$ e instável para $r>0$.

Nos equilíbrios $x_\pm^*$,

$$
f_x=r-3r=-2r<0.
$$

Logo, os dois novos equilíbrios são estáveis.

Quando $r$ atravessa zero, o estado central perde estabilidade e dá origem a dois estados estáveis: uma **pitchfork supercrítica**.

```python
rn=np.linspace(-3,0,300); rp=np.linspace(0,3,300)
plt.figure(figsize=(8,6))
plt.plot(rn,0*rn,linewidth=2)
plt.plot(rp,0*rp,"--",linewidth=2)
plt.plot(rp,np.sqrt(rp),linewidth=2)
plt.plot(rp,-np.sqrt(rp),linewidth=2)
plt.axhline(0,linewidth=.7); plt.axvline(0,linewidth=.7)
plt.xlabel("r"); plt.ylabel(r"$x^*$")
plt.title(r"Pitchfork supercrítica: $\dot{x}=rx-x^3$")
plt.grid(alpha=.25); plt.show()
```

## 7.8 O papel da simetria

Para

$$
f(x,r)=rx-x^3,
$$

vale

$$
f(-x,r)=-f(x,r).
$$

A equação é invariante pela reflexão $x\mapsto-x$. Por isso os dois novos equilíbrios surgem simetricamente.

Bifurcações pitchfork aparecem naturalmente em problemas nos quais um estado simétrico perde estabilidade e o sistema pode escolher entre dois estados equivalentes.

# 7.9 Pitchfork subcrítica

Considere agora

$$
\boxed{\dot{x}=rx+x^3=x(r+x^2).}
$$

Temos sempre $x^*=0$. Para $r<0$ existem também

$$
x_\pm^*=\pm\sqrt{-r}.
$$

Como

$$
f_x=r+3x^2,
$$

$x=0$ é estável para $r<0$ e instável para $r>0$.

Nos ramos não nulos,

$$
f_x=r+3(-r)=-2r>0,
$$

portanto eles são instáveis.

Essa é a **pitchfork subcrítica**.

```python
rn=np.linspace(-3,0,300); rp=np.linspace(0,3,300)
plt.figure(figsize=(8,6))
plt.plot(rn,0*rn,linewidth=2)
plt.plot(rp,0*rp,"--",linewidth=2)
plt.plot(rn,np.sqrt(-rn),"--",linewidth=2)
plt.plot(rn,-np.sqrt(-rn),"--",linewidth=2)
plt.axhline(0,linewidth=.7); plt.axvline(0,linewidth=.7)
plt.xlabel("r"); plt.ylabel(r"$x^*$")
plt.title(r"Pitchfork subcrítica: $\dot{x}=rx+x^3$")
plt.grid(alpha=.25); plt.show()
```

## 7.10 Um cuidado com a forma normal subcrítica

A equação

$$
\dot{x}=rx+x^3
$$

descreve o comportamento **local** perto de $(x,r)=(0,0)$.

Em uma aplicação, termos de ordem superior podem alterar a dinâmica longe da origem. Por exemplo,

$$
\boxed{\dot{x}=rx+x^3-x^5}
$$

mantém a estrutura local da pitchfork subcrítica, mas pode produzir outros equilíbrios e regiões de coexistência de estados estáveis.

Essa observação será importante no laboratório sobre transições abruptas e histerese.

# 7.11 Desaceleração crítica

Na pitchfork supercrítica,

$$
\dot{x}=rx-x^3.
$$

Para $r<0$, $x^*=0$ é estável. Perto da origem,

$$
\dot{x}\approx rx,
$$

e portanto

$$
x(t)\approx x_0e^{rt}.
$$

A escala de tempo característica é

$$
\boxed{\tau=\frac1{|r|}.}
$$

Quando $r\to0^-$,

$$
\tau\to\infty.
$$

O retorno ao equilíbrio fica progressivamente mais lento ao nos aproximarmos da bifurcação. Esse fenômeno é chamado de **desaceleração crítica** (*critical slowing down*).

```python
from scipy.integrate import solve_ivp

def sol_pitchfork(r,x0=.8,T=30):
    sol=solve_ivp(lambda t,X:r*X[0]-X[0]**3,[0,T],[x0],
                  t_eval=np.linspace(0,T,1000))
    return sol.t,sol.y[0]

plt.figure(figsize=(9,5))
for r in [-1,-.5,-.2,-.05]:
    t,x=sol_pitchfork(r)
    plt.plot(t,x,label=fr"$r={r}$")
plt.xlabel("t"); plt.ylabel("x(t)")
plt.title("Desaceleração crítica")
plt.grid(alpha=.25); plt.legend(); plt.show()
```

# 7.12 Explorador interativo

Use o controle abaixo para observar como o gráfico de $f(x,r)$ muda quando $r$ atravessa o valor crítico.

```python
from ipywidgets import interact, FloatSlider, Dropdown

def explorador(tipo="sela-nó",r=-1):
    if tipo=="sela-nó":
        f=lambda x:r+x**2; titulo=r"$\dot{x}=r+x^2$"
    elif tipo=="transcrítica":
        f=lambda x:r*x-x**2; titulo=r"$\dot{x}=rx-x^2$"
    elif tipo=="pitchfork supercrítica":
        f=lambda x:r*x-x**3; titulo=r"$\dot{x}=rx-x^3$"
    else:
        f=lambda x:r*x+x**3; titulo=r"$\dot{x}=rx+x^3$"

    x=np.linspace(-3,3,1000)
    plt.figure(figsize=(9,5))
    plt.plot(x,f(x)); plt.axhline(0,linewidth=.9); plt.axvline(0,linewidth=.7)
    plt.xlabel("x"); plt.ylabel(r"$\dot{x}$")
    plt.title(titulo+fr",  $r={r:.2f}$")
    plt.ylim(-6,6); plt.grid(alpha=.25); plt.show()

interact(explorador,
         tipo=Dropdown(options=["sela-nó","transcrítica",
                                "pitchfork supercrítica","pitchfork subcrítica"]),
         r=FloatSlider(value=-1,min=-2,max=2,step=.05));
```

# 7.13 Roteiro para construir um diagrama de bifurcação

Para

$$
\dot{x}=f(x,r):
$$

1. Resolva $f(x,r)=0$ e determine os ramos $x^*(r)$.
2. Identifique valores de $r$ em que o número de equilíbrios ou sua estabilidade muda.
3. Calcule $f_x(x^*,r)$ quando possível.
4. Classifique os ramos como estáveis ou instáveis.
5. Represente $x^*$ em função de $r$.
6. Descreva a mudança qualitativa que ocorre ao atravessar cada valor crítico.

Para uma sela-nó genérica, as condições

$$
f=0,\qquad f_x=0
$$

são particularmente úteis para localizar o ponto crítico.

# 7.14 Comparação das formas normais

| Bifurcação | Forma normal | Comportamento |
|---|---|---|
| Sela-nó | $\dot{x}=r+x^2$ | dois equilíbrios são criados/destruídos |
| Transcrítica | $\dot{x}=rx-x^2$ | dois ramos trocam estabilidade |
| Pitchfork supercrítica | $\dot{x}=rx-x^3$ | surgem dois novos ramos estáveis |
| Pitchfork subcrítica | $\dot{x}=rx+x^3$ | ramos instáveis encontram o ramo central |

Essas equações simples representam padrões locais que reaparecem em muitos modelos diferentes.
