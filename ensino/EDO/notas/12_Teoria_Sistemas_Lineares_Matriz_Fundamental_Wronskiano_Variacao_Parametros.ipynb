> **Qual é a estrutura do conjunto de todas as soluções do sistema linear $\mathbf{x}'=A\mathbf{x}$?**

Estudaremos sistemas $\mathbf{x}'=A(t)\mathbf{x}$, espaço de soluções, independência linear, matriz fundamental, Wronskiano, fórmula de Liouville, matriz de transição e sistemas não homogêneos por variação dos parâmetros.


# 1. O problema linear geral

Considere
$$
\boxed{\mathbf{x}'(t)=A(t) \, \, \mathbf{x}(t)},
$$
onde $A(t)$ é uma matriz $n\times n$ com entradas contínuas em um intervalo $I$ e $\mathbf{x}:I\to\mathbb R^n$.

O caso $\mathbf{x}'=A\mathbf{x}$ das aulas anteriores corresponde a $A(t)$ constante.

Para cada $t_0\in I$ e $\mathbf{x}_0\in\mathbb R^n$, existência e unicidade fornecem uma única solução de
$$
\mathbf{x}'=A(t)\mathbf{x},\qquad \mathbf{x}(t_0)=\mathbf{x}_0.
$$

Veremos que essa unicidade terá consequências algébricas profundas.

# 2. Linearidade e princípio da superposição

Defina o operador
$$
L[\mathbf{x}]=\mathbf{x}'-A(t)\mathbf{x}.
$$
Observe que
$$
L[\alpha\mathbf{x}+\beta\mathbf{y}]
=\alpha L[\mathbf{x}]+\beta L[\mathbf{y}].
$$
**Qual o domínio desse operador?**

Observe ainda que, se $\mathbf{x}_1,\ldots,\mathbf{x}_k$ são soluções do sistema homogêneo, então
$$
\sum_{j=1}^k c_j\mathbf{x}_j
$$
também é solução.


# 3. O espaço de soluções

Defina
$$
\mathcal S=\{\mathbf{x}\in C^1(I,\mathbb R^n):\mathbf{x}'=A(t)\mathbf{x}\}.
$$

Pela superposição, $\mathcal S$ é um espaço vetorial.

Mas seus elementos são funções. Qual é sua dimensão?

Fixe $t_0\in I$ e defina o mapa de avaliação
$$
E_{t_0}:\mathcal S\to\mathbb R^n,\qquad E_{t_0}(\mathbf{x})=\mathbf{x}(t_0).
$$

Ele é linear. É sobrejetivo porque, para qualquer $\mathbf{x}_0$, existe uma solução com $\mathbf{x}(t_0)=\mathbf{x}_0$. É injetivo porque duas soluções com o mesmo valor em $t_0$ resolvem o mesmo PVI e, pela unicidade, coincidem.

Portanto,
$$
\boxed{E_{t_0}:\mathcal S\to\mathbb R^n\text{ é um isomorfismo}.}
$$
Consequentemente,
$$
\boxed{\dim\mathcal S=n.}
$$

Isso explica por que a solução geral envolve $n$ constantes arbitrárias.

# 4. Independência linear de soluções

Soluções $\mathbf{x}_1,\ldots,\mathbf{x}_n$ são linearmente independentes se
$$
c_1\mathbf{x}_1(t)+\cdots+c_n\mathbf{x}_n(t)\equiv0
$$
implica $c_1=\cdots=c_n=0$.

**Teorema.** Elas são independentes em $I$ se, e somente se, os vetores
$$
\mathbf{x}_1(t_0),\ldots,\mathbf{x}_n(t_0)
$$
são independentes em $\mathbb R^n$ em um único instante $t_0$.

Para a direção menos imediata, suponha
$$
c_1\mathbf{x}_1(t_0)+\cdots+c_n\mathbf{x}_n(t_0)=0.
$$
A função
$$
\mathbf z(t)=\sum_jc_j\mathbf{x}_j(t)
$$
é solução e satisfaz $\mathbf z(t_0)=0$. Pela unicidade do PVI, $\mathbf z\equiv0$.

Assim, uma informação em **um instante** se propaga para **todo o intervalo**.

# 5. Matriz fundamental

Se $\mathbf{x}_1,\ldots,\mathbf{x}_n$ formam uma base de $\mathcal S$, definimos
$$
\boxed{
X(t)=
\begin{pmatrix}
|&&|\\
\mathbf{x}_1(t)&\cdots&\mathbf{x}_n(t)\\
|&&|
\end{pmatrix}.
}
$$

Chamamos $X(t)$ de **matriz fundamental**. Como cada coluna resolve o sistema,
$$
\boxed{X'(t)=A(t)X(t).}
$$

A solução geral fica
$$
\boxed{\mathbf{x}(t)=X(t)\mathbf c,\qquad \mathbf c\in\mathbb R^n.}
$$

# 6. Wronskiano

Definimos
$$
\boxed{W(t)=\det X(t).}
$$

Como as colunas de $X(t)$ são os vetores solução,
$$
W(t_0)\neq0
$$
se, e somente se, eles são independentes em $t_0$. Pelo teorema anterior,
$$
\boxed{W(t_0)\neq0\Longrightarrow W(t)\neq0\quad\forall t\in I.}
$$

Para soluções de um mesmo sistema homogêneo existe, portanto, a dicotomia:
$$
\boxed{W(t)\neq0\ \forall t}
\qquad\text{ou}\qquad
\boxed{W(t)\equiv0.}
$$

# 7. Exemplo

Para
$$
\mathbf{x}'=
\begin{pmatrix}1&0\\0&-2\end{pmatrix}\mathbf{x},
$$
podemos escolher
$$
\mathbf{x}_1=e^t\binom10,\qquad
\mathbf{x}_2=e^{-2t}\binom01.
$$
Então
$$
X(t)=
\begin{pmatrix}e^t&0\\0&e^{-2t}\end{pmatrix},
\qquad
W(t)=e^{-t}\neq0.
$$
Logo as duas soluções formam uma base de $\mathcal S$.

# 8. Fórmula de Liouville

Se $X'=A(t)X$, então
$$
\boxed{W'(t)=\operatorname{tr}(A(t))W(t).}
$$
Logo
$$
\boxed{
W(t)=W(t_0)\exp\left(\int_{t_0}^{t}\operatorname{tr}(A(s))\,ds\right).
}
$$

Para demonstrar, usamos a fórmula de Jacobi:
$$
\frac{d}{dt}\det X=\det X\,\operatorname{tr}(X^{-1}X').
$$
Como $X'=AX$,
$$
X^{-1}X'=X^{-1}AX,
$$
e o traço é invariante por similaridade:
$$
\operatorname{tr}(X^{-1}AX)=\operatorname{tr}(A).
$$

A exponencial nunca se anula; portanto uma família fundamental nunca perde sua independência.

## 8.1 Interpretação geométrica

Em dimensão dois, $|\det X(t)|$ é a área do paralelogramo formado pelas colunas de $X(t)$. Em dimensão três, é um volume.

Assim,
$$
\frac{W'(t)}{W(t)}=\operatorname{tr}A(t)
$$
mostra que o traço mede a taxa instantânea de expansão ou contração de volumes pelo fluxo linear.

Isso dá uma nova interpretação ao traço estudado na Aula 11.

```python
import numpy as np
import matplotlib.pyplot as plt

t=np.linspace(0,3,300)
W=np.exp(-t)

plt.figure(figsize=(7,4))
plt.plot(t,W)
plt.xlabel("t")
plt.ylabel("W(t)")
plt.title(r"Exemplo: $W(t)=e^{-t}$")
plt.grid(alpha=.25)
plt.show()
```

# 9. Matriz fundamental principal

Fixado $t_0$, chamamos de matriz fundamental principal $\Phi(t,t_0)$ a solução matricial de
$$
\frac{\partial}{\partial t}\Phi(t,t_0)=A(t)\Phi(t,t_0),
\qquad
\boxed{\Phi(t_0,t_0)=I.}
$$

Suas colunas são as soluções dos PVIs
$$
\mathbf{x}_j'=A(t)\mathbf{x}_j,\qquad
\mathbf{x}_j(t_0)=\mathbf e_j.
$$

Assim, resolver uma equação matricial com dado inicial $I$ equivale a resolver simultaneamente os PVIs correspondentes a uma base de $\mathbb R^n$.

# 10. De uma matriz fundamental a qualquer PVI

Se
$$
\mathbf{x}(t)=X(t)\mathbf c
$$
e $\mathbf{x}(t_0)=\mathbf{x}_0$, então
$$
\mathbf c=X(t_0)^{-1}\mathbf{x}_0.
$$
Logo
$$
\boxed{\mathbf{x}(t)=X(t)X(t_0)^{-1}\mathbf{x}_0.}
$$

Definimos
$$
\boxed{\Phi(t,s)=X(t)X(s)^{-1}.}
$$
Então
$$
\boxed{\mathbf{x}(t)=\Phi(t,t_0)\mathbf{x}_0.}
$$

A matriz $\Phi(t,s)$ transporta o estado no instante $s$ para o estado no instante $t$.

# 11. Propriedades da matriz de transição

Diretamente da definição,
$$
\boxed{\Phi(s,s)=I,}
$$
$$
\boxed{\Phi(t,s)\Phi(s,r)=\Phi(t,r),}
$$
e
$$
\boxed{\Phi(t,s)^{-1}=\Phi(s,t).}
$$

A segunda identidade traduz a consistência da evolução temporal: evoluir de $r$ para $s$ e depois de $s$ para $t$ é o mesmo que evoluir diretamente de $r$ para $t$.

# 12. Duas matrizes fundamentais

Se $X(t)$ e $Y(t)$ são matrizes fundamentais do mesmo sistema, então
$$
\boxed{Y(t)=X(t)C}
$$
para alguma matriz constante invertível $C$.

De fato, as colunas de $X$ formam uma base do espaço de soluções; portanto cada coluna de $Y$ é combinação linear, com coeficientes constantes, das colunas de $X$.

Além disso,
$$
\det Y=\det X\,\det C\neq0,
$$
logo $\det C\neq0$.

Portanto matrizes fundamentais diferentes são apenas diferentes escolhas de base para o mesmo espaço de soluções.

# 13. A matriz de transição é intrínseca

Se $Y(t)=X(t)C$, então
$$
Y(t)Y(s)^{-1}
=X(t)C(X(s)C)^{-1}
=X(t)X(s)^{-1}.
$$

Assim, embora $X(t)$ não seja única,
$$
\boxed{\Phi(t,s)}
$$
não depende da base escolhida.

Essa distinção é importante:

- **matriz fundamental:** depende da escolha de uma base de soluções;
- **matriz de transição:** pertence ao próprio sistema.

# 14. Coeficientes constantes e $e^{At}$

Se $A(t)\equiv A$, então
$$
X(t)=e^{At}
$$
é matriz fundamental, pois
$$
\frac{d}{dt}e^{At}=Ae^{At},\qquad e^{A0}=I.
$$

Consequentemente,
$$
\boxed{\Phi(t,s)=e^{A(t-s)}}
$$
e
$$
\boxed{\mathbf{x}(t)=e^{A(t-t_0)}\mathbf{x}_0.}
$$

Portanto a exponencial matricial estudada anteriormente é a matriz de evolução do sistema autônomo linear.

# 15. Exemplo completo

Considere
$$
\mathbf{x}'=
\begin{pmatrix}0&1\\-2&-3\end{pmatrix}\mathbf{x}.
$$

Os autovalores são $-1$ e $-2$, com autovetores
$$
\mathbf v_1=\binom{1}{-1},
\qquad
\mathbf v_2=\binom{1}{-2}.
$$

Uma matriz fundamental é
$$
X(t)=
\begin{pmatrix}
e^{-t}&e^{-2t}\\
-e^{-t}&-2e^{-2t}
\end{pmatrix}.
$$

Logo
$$
W(t)=-e^{-3t}.
$$

Como $\operatorname{tr}A=-3$, Liouville fornece
$$
W(t)=W(0)e^{-3t}=-e^{-3t},
$$
confirmando o cálculo.

## 15.1 Um PVI

Para
$$
\mathbf{x}(0)=\binom{2}{-3},
$$
escrevemos $\mathbf{x}=X(t)\mathbf c$. Como
$$
X(0)=\begin{pmatrix}1&1\\-1&-2\end{pmatrix},
$$
obtemos
$$
\begin{pmatrix}1&1\\-1&-2\end{pmatrix}\binom{c_1}{c_2}
=\binom{2}{-3}.
$$
Daí $c_1=c_2=1$ e
$$
\boxed{
\mathbf{x}(t)
=e^{-t}\binom{1}{-1}
+e^{-2t}\binom{1}{-2}.
}
$$

# 16. Sistemas não homogêneos

Agora considere
$$
\boxed{\mathbf{x}'=A(t)\mathbf{x}+\mathbf f(t).}
$$

O conjunto de soluções já não é, em geral, um espaço vetorial.

Se $\mathbf{x}_p$ é uma solução particular e $\mathbf{x}_h$ resolve o sistema homogêneo, então
$$
\mathbf{x}_p+\mathbf{x}_h
$$
é solução não homogênea.

Reciprocamente, a diferença de duas soluções não homogêneas resolve o sistema homogêneo.

Logo
$$
\boxed{\mathbf{x}=\mathbf{x}_p+\mathbf{x}_h.}
$$

O conjunto de soluções não homogêneas é uma translação afim do espaço de soluções homogêneas.

# 17. A ideia da variação dos parâmetros

No sistema homogêneo,
$$
\mathbf{x}_h(t)=X(t)\mathbf c,
$$
com $\mathbf c$ constante.

Para o sistema não homogêneo, substituímos a constante por uma função:
$$
\boxed{\mathbf{x}(t)=X(t)\mathbf c(t).}
$$

Derivando,
$$
\mathbf{x}'=X'\mathbf c+X\mathbf c'.
$$
Como $X'=AX$,
$$
\mathbf{x}'=AX\mathbf c+X\mathbf c'.
$$
Comparando com
$$
\mathbf{x}'=AX\mathbf c+\mathbf f,
$$
temos
$$
X(t)\mathbf c'(t)=\mathbf f(t).
$$
Como $X(t)$ é invertível,
$$
\boxed{\mathbf c'(t)=X(t)^{-1}\mathbf f(t).}
$$

# 18. Fórmula de variação dos parâmetros

Integrando,
$$
\mathbf c(t)=\mathbf c(t_0)+
\int_{t_0}^{t}X(s)^{-1}\mathbf f(s)\,ds.
$$

Usando o dado inicial,
$$
\mathbf c(t_0)=X(t_0)^{-1}\mathbf{x}_0.
$$

Portanto,
$$
\boxed{
\mathbf{x}(t)
=X(t)X(t_0)^{-1}\mathbf{x}_0
+X(t)\int_{t_0}^{t}X(s)^{-1}\mathbf f(s)\,ds.
}
$$

Como $\Phi(t,s)=X(t)X(s)^{-1}$,
$$
\boxed{
\mathbf{x}(t)
=
\Phi(t,t_0)\mathbf{x}_0
+
\int_{t_0}^{t}\Phi(t,s)\mathbf f(s)\,ds.
}
$$

A primeira parcela é a evolução livre do dado inicial. A segunda acumula as entradas externas aplicadas em cada instante $s$ e transportadas até $t$.

# 19. Caso $A$ constante

Se $A$ é constante,
$$
\Phi(t,s)=e^{A(t-s)}.
$$
Logo
$$
\boxed{
\mathbf{x}(t)
=
e^{A(t-t_0)}\mathbf{x}_0
+
\int_{t_0}^{t}e^{A(t-s)}\mathbf f(s)\,ds.
}
$$

Essa expressão é frequentemente chamada de **fórmula de variação das constantes** ou **fórmula de Duhamel**.

Ela é o análogo matricial do método do fator integrante para EDOs lineares escalares.

# 20. Conexão com o fator integrante

Para
$$
y'=a(t)y+f(t),
$$
uma solução homogênea é
$$
X(t)=\exp\left(\int_{t_0}^{t}a(r)\,dr\right).
$$

A fórmula matricial reduz-se a
$$
y(t)=X(t)X(t_0)^{-1}y_0+
X(t)\int_{t_0}^{t}X(s)^{-1}f(s)\,ds,
$$
que é exatamente a fórmula obtida pelo fator integrante.

Portanto:
$$
\boxed{\text{fator integrante escalar}\subset\text{variação dos parâmetros para sistemas}.}
$$

# 21. Exemplo de variação dos parâmetros

Considere
$$
\begin{cases}
x'=x+e^t,\\
y'=-y+1,
\end{cases}
\qquad
\mathbf{x}(0)=\binom00.
$$

Temos
$$
A=\begin{pmatrix}1&0\\0&-1\end{pmatrix},
\qquad
\mathbf f(t)=\binom{e^t}{1},
$$
e
$$
X(t)=\begin{pmatrix}e^t&0\\0&e^{-t}\end{pmatrix},
\qquad
X(t)^{-1}=\begin{pmatrix}e^{-t}&0\\0&e^t\end{pmatrix}.
$$

Assim,
$$
X(s)^{-1}\mathbf f(s)=\binom{1}{e^s},
$$
e
$$
\int_0^tX(s)^{-1}\mathbf f(s)\,ds
=\binom{t}{e^t-1}.
$$

Logo
$$
\boxed{
\mathbf{x}(t)=
\binom{te^t}{1-e^{-t}}.
}
$$

```python
t=np.linspace(0,3,300)
x=t*np.exp(t)
y=1-np.exp(-t)

plt.figure(figsize=(8,4.5))
plt.plot(t,x,label="x(t)")
plt.plot(t,y,label="y(t)")
plt.xlabel("t")
plt.ylabel("componentes")
plt.title("Sistema linear não homogêneo")
plt.legend()
plt.grid(alpha=.25)
plt.show()
```

# 22. EDOs escalares de segunda ordem como sistemas

Considere
$$
y''+p(t)y'+q(t)y=0.
$$
Com
$$
x_1=y,\qquad x_2=y',
$$
obtemos
$$
\binom{x_1}{x_2}'
=
\begin{pmatrix}
0&1\\
-q(t)&-p(t)
\end{pmatrix}
\binom{x_1}{x_2}.
$$

Se $y_1,y_2$ são soluções, uma matriz fundamental é
$$
X(t)=
\begin{pmatrix}
y_1&y_2\\
y_1'&y_2'
\end{pmatrix}.
$$

Logo
$$
\boxed{W(t)=y_1y_2'-y_1'y_2,}
$$
o Wronskiano clássico das equações escalares de segunda ordem.

# 23. Abel como caso particular de Liouville

Para o sistema anterior,
$$
A(t)=
\begin{pmatrix}
0&1\\
-q(t)&-p(t)
\end{pmatrix},
$$
e
$$
\operatorname{tr}A(t)=-p(t).
$$

Liouville fornece
$$
\boxed{
W(t)=W(t_0)
\exp\left(-\int_{t_0}^{t}p(s)\,ds\right).
}
$$

Essa é a **fórmula de Abel**.

Portanto:
$$
\boxed{\text{Abel para EDO escalar}\subset\text{Liouville para sistemas}.}
$$

# 24. Um cuidado sobre o Wronskiano

Para funções arbitrárias, $W(t_0)=0$ em um ponto não implica, por si só, dependência linear.

O resultado forte desta aula vale porque tratamos de soluções de uma **mesma EDO linear homogênea**, sob hipóteses que garantem unicidade.

Novamente, a teoria de existência e unicidade aparece por trás de um resultado aparentemente apenas algébrico.
