# 5. Equações Lineares de Primeira Ordem

Nesta seção estudaremos uma das classes mais importantes de equações diferenciais de primeira ordem: as **equações lineares**.

Seguiremos de perto a abordagem da Lição 11 de Tenenbaum e Pollard. A ideia central será reconhecer a forma linear, determinar um **fator integrante**, transformar a equação em uma derivada de um produto e, ao final, estudar as equações de **Bernoulli**.



## 5.1 Definição

Uma equação diferencial linear de primeira ordem é uma equação que pode ser escrita na forma

$$
\boxed{y'+P(x)y=Q(x),}
$$

onde $P$ e $Q$ são contínuas no intervalo em que procuramos as soluções.

A palavra **linear** significa que $y$ e $y'$ aparecem com potência $1$ e não aparecem multiplicados entre si.

Por exemplo,

$$
y'+3xy=\sin x
$$

é linear, enquanto

$$
yy'+x=0
$$

e

$$
y'+y^2=x
$$

não são lineares.

## 5.2 A ideia do fator integrante

Considere

$$
y'+P(x)y=Q(x).
$$

Multiplicamos por uma função $\mu(x)$:

$$
\mu y'+\mu Py=\mu Q.
$$

Queremos que o lado esquerdo seja

$$
\frac{d}{dx}(\mu y).
$$

Pela regra do produto,

$$
(\mu y)'=\mu y'+\mu' y.
$$

Logo, precisamos de

$$
\mu'=\mu P.
$$

Assim,

$$
\frac{\mu'}{\mu}=P(x),
$$

e, integrando,

$$
\ln|\mu|=\int P(x)\,dx.
$$

Podemos escolher

$$
\boxed{\mu(x)=e^{\int P(x)\,dx}.}
$$

Essa função é chamada de **fator integrante**.

## 5.3 A dedução pela forma diferencial

Observem que o fator integrante pode ser relacionado ao conceito de equação exata.

Da equação

$$
y'+P(x)y=Q(x)
$$

obtemos

$$
\big(P(x)y-Q(x)\big)\,dx+dy=0.
$$

Multiplicando por $\mu(x)$,

$$
\mu(x)\big(P(x)y-Q(x)\big)\,dx+\mu(x)\,dy=0.
$$

Para que essa equação seja exata, devemos ter

$$
\frac{\partial}{\partial y}
\left[\mu P y-\mu Q\right]
=
\frac{\partial}{\partial x}\mu.
$$

Portanto,

$$
\mu P=\mu',
$$

isto é,

$$
\frac{\mu'}{\mu}=P.
$$

Integrando,

$$
\boxed{\mu=e^{\int P(x)\,dx}.}
$$

Portanto, o fator integrante é escolhido para tornar a equação exata.

## 5.4 A derivada de um produto

Multiplicando a equação por $\mu$,

$$
\mu y'+\mu Py=\mu Q.
$$

Como $\mu'=\mu P$,

$$
\mu y'+\mu' y=\mu Q.
$$

Logo,

$$
\boxed{(\mu y)'=\mu Q.}
$$

Integramos:

$$
\mu(x)y(x)=\int \mu(x)Q(x)\,dx+C.
$$

Portanto,

$$
\boxed{
y(x)=
\frac1{\mu(x)}
\left[
\int \mu(x)Q(x)\,dx+C
\right].
}
$$

Como

$$
\mu=e^{\int P\,dx},
$$

obtemos

$$
\boxed{
y(x)=
e^{-\int P\,dx}
\left[
\int e^{\int P\,dx}Q(x)\,dx+C
\right].
}
$$

## 5.5 Por que a solução é geral?

Como

$$
\mu(x)=e^{\int P(x)\,dx}>0,
$$

o fator integrante nunca se anula.

Assim, cada transformação

$$
y'+Py=Q
\Longleftrightarrow
\mu y'+\mu Py=\mu Q
\Longleftrightarrow
(\mu y)'=\mu Q
$$

é reversível.

Portanto, toda solução da equação original aparece na família obtida após a integração. Isso justifica chamar essa família de **solução geral**.

## 5.6 Exemplo 1

Resolva

$$
y'-2xy=e^{x^2}.
$$

Temos

$$
P(x)=-2x,
\qquad
Q(x)=e^{x^2}.
$$

Logo,

$$
\mu(x)=e^{\int -2x\,dx}=e^{-x^2}.
$$

Multiplicando,

$$
e^{-x^2}y'-2xe^{-x^2}y=1.
$$

Mas

$$
\left(e^{-x^2}y\right)'
=
e^{-x^2}y'-2xe^{-x^2}y.
$$

Portanto,

$$
\left(e^{-x^2}y\right)'=1.
$$

Integrando,

$$
e^{-x^2}y=x+C,
$$

e

$$
\boxed{y=e^{x^2}(x+C).}
$$

## 5.7 Verificação simbólica

Vamos usar o computador apenas para verificar a expressão encontrada.

```python
import sympy as sp

x, C = sp.symbols('x C')

y = sp.exp(x**2)*(x + C)

sp.simplify(sp.diff(y, x) - 2*x*y - sp.exp(x**2))
```

O resultado deve ser `0`. Isso confirma simbolicamente que

$$
y=e^{x^2}(x+C)
$$

satisfaz a equação para todo valor da constante $C$.

## 5.8 Visualizando a família de soluções

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1.5, 1.5, 500)

plt.figure(figsize=(8,6))

for C in [-2, -1, 0, 1, 2]:
    y = np.exp(x**2)*(x + C)
    plt.plot(x, y, label=fr"$C={C}$")

plt.axhline(0, linewidth=.8)
plt.axvline(0, linewidth=.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Soluções de $y'-2xy=e^{x^2}$")
plt.ylim(-15,15)
plt.grid(alpha=.25)
plt.legend()
plt.show()
```

## 5.9 A equação homogênea associada

À equação

$$
y'+P(x)y=Q(x)
$$

associamos

$$
\boxed{y'+P(x)y=0.}
$$

Essa equação é separável:

$$
\frac{dy}{dx}=-P(x)y.
$$

Logo,

$$
\frac{dy}{y}=-P(x)\,dx,
$$

e

$$
\boxed{y_h=Ce^{-\int P(x)\,dx}.}
$$

Assim, a solução geral da equação não homogênea pode ser escrita como

$$
\boxed{y=y_p+y_h,}
$$

onde

$$
y_h=Ce^{-\int P\,dx}
$$

e uma solução particular pode ser tomada como

$$
y_p=
e^{-\int P\,dx}
\int e^{\int P\,dx}Q(x)\,dx.
$$

## 5.10 Exemplo 2 — problema de valor inicial

Considere

$$
x\frac{dy}{dx}+3y=\sin x,
\qquad x\neq0,
$$

com

$$
y\left(\frac{\pi}{2}\right)=1.
$$

Dividindo por $x$,

$$
y'+\frac3x y=\frac{\sin x}{x}.
$$

Logo,

$$
P(x)=\frac3x,
\qquad
Q(x)=\frac{\sin x}{x}.
$$

Como o ponto inicial está em $x>0$, trabalhamos no intervalo $(0,\infty)$.

O fator integrante é

$$
\mu(x)
=
e^{\int 3/x\,dx}
=
x^3.
$$

Multiplicando,

$$
x^3y'+3x^2y=x^2\sin x,
$$

isto é,

$$
\boxed{(x^3y)'=x^2\sin x.}
$$

Integrando,

$$
x^3y
=
\int x^2\sin x\,dx+C.
$$

Por integração por partes,

$$
\int x^2\sin x\,dx
=
-x^2\cos x+2x\sin x+2\cos x.
$$

Portanto,

$$
x^3y
=
-x^2\cos x+2x\sin x+2\cos x+C.
$$

Usando $y(\pi/2)=1$,

$$
C=\frac{\pi^3}{8}-\pi.
$$

Assim,

$$
\boxed{
y(x)=
\frac{-x^2\cos x+2x\sin x+2\cos x+\frac{\pi^3}{8}-\pi}{x^3},
\qquad x>0.
}
$$

```python
x = np.linspace(0.35, 8, 700)

C0 = np.pi**3/8 - np.pi
y = (-x**2*np.cos(x) + 2*x*np.sin(x) + 2*np.cos(x) + C0)/x**3

plt.figure(figsize=(8,5))
plt.plot(x, y)
plt.scatter([np.pi/2], [1], s=50)
plt.axhline(0, linewidth=.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Solução do PVI $xy'+3y=\sin x$")
plt.grid(alpha=.25)
plt.show()
```

## 5.11 O fator integrante como transformação

Considere novamente

$$
y'-2xy=e^{x^2}.
$$

Defina

$$
z=e^{-x^2}y.
$$

Como o fator integrante é $\mu=e^{-x^2}$, a equação transforma-se em

$$
z'=1.
$$

Assim,

$$
z=x+C.
$$

O fator integrante transforma uma família de soluções relativamente complicada em uma família de retas paralelas.

```python
x = np.linspace(-1.5, 1.5, 500)

plt.figure(figsize=(8,5))

for C in [-2, -1, 0, 1, 2]:
    y = np.exp(x**2)*(x+C)
    z = np.exp(-x**2)*y
    plt.plot(x, z, label=fr"$C={C}$")

plt.xlabel("x")
plt.ylabel(r"$z=e^{-x^2}y$")
plt.title(r"Após o fator integrante: $z'=1$")
plt.grid(alpha=.25)
plt.legend()
plt.show()
```

# 5.12 Equação de Bernoulli

Uma equação de Bernoulli tem a forma

$$
\boxed{
y'+P(x)y=Q(x)y^n.
}
$$

Se $n=0$, ela já é linear.

Se $n=1$,

$$
y'=[Q(x)-P(x)]y,
$$

e a equação é separável.

O caso realmente novo é

$$
n\neq0,1.
$$

## 5.13 A transformação de Bernoulli

Multiplicamos

$$
y'+P(x)y=Q(x)y^n
$$

por

$$
(1-n)y^{-n}.
$$

Obtemos

$$
(1-n)y^{-n}y'
+
(1-n)P(x)y^{1-n}
=
(1-n)Q(x).
$$

Pela regra da cadeia,

$$
\frac{d}{dx}\left(y^{1-n}\right)
=
(1-n)y^{-n}y'.
$$

Faça então

$$
\boxed{u=y^{1-n}.}
$$

A equação torna-se

$$
\boxed{
u'+(1-n)P(x)u=(1-n)Q(x),
}
$$

que é linear em $u$.

## 5.14 Exemplo de Bernoulli

Considere

$$
y'+xy=xy^{-3},
\qquad y\neq0.
$$

Aqui,

$$
P(x)=x,
\qquad
Q(x)=x,
\qquad
n=-3.
$$

Como $1-n=4$, multiplicamos por $4y^3$:

$$
4y^3y'+4xy^4=4x.
$$

Como

$$
(y^4)'=4y^3y',
$$

obtemos

$$
(y^4)'+4xy^4=4x.
$$

Definindo

$$
u=y^4,
$$

temos a equação linear

$$
u'+4xu=4x.
$$

O fator integrante é

$$
\mu=e^{\int4x\,dx}=e^{2x^2}.
$$

Portanto,

$$
(e^{2x^2}u)'=4xe^{2x^2}.
$$

Integrando,

$$
e^{2x^2}u=e^{2x^2}+C,
$$

logo

$$
u=1+Ce^{-2x^2}.
$$

Como $u=y^4$,

$$
\boxed{
y^4=1+Ce^{-2x^2}.
}
$$

## 5.15 Visualizando as soluções de Bernoulli

Vamos representar o ramo positivo das soluções reais.

```python
x = np.linspace(-2, 2, 600)

plt.figure(figsize=(8,5))

for C in [-0.8, -0.3, 0, 1, 3]:
    rad = 1 + C*np.exp(-2*x**2)
    mask = rad > 0
    y = np.full_like(x, np.nan)
    y[mask] = rad[mask]**0.25
    plt.plot(x, y, label=fr"$C={C}$")

plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Ramo positivo de $y^4=1+Ce^{-2x^2}$")
plt.grid(alpha=.25)
plt.legend()
plt.show()
```

## 5.16 Roteiro para equações lineares

Para

$$
y'+P(x)y=Q(x):
$$

1. Coloque a equação na forma padrão.
2. Identifique um intervalo onde $P$ e $Q$ sejam contínuas.
3. Calcule
   $$
   \mu=e^{\int P(x)\,dx}.
   $$
4. Multiplique a equação por $\mu$.
5. Reconheça
   $$
   (\mu y)'=\mu Q.
   $$
6. Integre:
   $$
   \mu y=\int \mu Q\,dx+C.
   $$
7. Resolva para $y$.
8. Se houver condição inicial, determine $C$ e escolha o intervalo apropriado.

## 5.17 Roteiro para Bernoulli

Para

$$
y'+P(x)y=Q(x)y^n,
\qquad n\neq0,1,
$$

use

$$
u=y^{1-n}.
$$

A equação transforma-se em

$$
u'+(1-n)P(x)u=(1-n)Q(x).
$$

Resolva a equação linear em $u$ e depois retorne a $y$.

Como em outras mudanças de variável, examine separadamente soluções que possam ter sido excluídas por divisões envolvendo potências de $y$.
