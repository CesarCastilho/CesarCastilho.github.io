## 1. De uma EDO escalar para um sistema

Um sistema bidimensional geral tem a forma

$$
\begin{cases}
x'=f(t,x,y),\\
y'=g(t,x,y).
\end{cases}
$$

Definindo

$$
\mathbf{x}(t)=\begin{pmatrix}x(t)\\y(t)\end{pmatrix},
\qquad
\mathbf{F}(t,\mathbf{x})=
\begin{pmatrix}f(t,x,y)\\g(t,x,y)\end{pmatrix},
$$

escrevemos

$$
\boxed{\mathbf{x}'=\mathbf{F}(t,\mathbf{x}).}
$$

Em dimensão $n$,

$$
\mathbf F:D\subset\mathbb R\times\mathbb R^n\to\mathbb R^n.
$$

Assim, um sistema de $n$ EDOs de primeira ordem é uma única EDO cujo estado pertence a $\mathbb R^n$.


## 2. Problema de valor inicial

Um PVI para um sistema é

$$
\begin{cases}
\mathbf{x}'(t)=\mathbf F(t,\mathbf{x}(t)),\\
\mathbf{x}(t_0)=\mathbf{x}_0.
\end{cases}
$$

Uma solução é uma função vetorial diferenciável

$$
\mathbf{x}:I\to\mathbb R^n
$$

que satisfaz simultaneamente todas as equações e a condição inicial.


## 3. EDOs de ordem superior como sistemas

Considere

$$
y''=F(t,y,y').
$$

Defina

$$
x_1=y,\qquad x_2=y'.
$$

Então

$$
\boxed{
\begin{cases}
x_1'=x_2,\\
x_2'=F(t,x_1,x_2).
\end{cases}}
$$

Portanto, uma EDO de segunda ordem pode ser estudada como um sistema bidimensional de primeira ordem.

### Exemplo — Oscilador harmônico

Para

$$
y''+\omega^2y=0,
$$

temos

$$
\boxed{
\begin{cases}
x_1'=x_2,\\
x_2'=-\omega^2x_1,
\end{cases}}
$$

ou

$$
\mathbf{x}'=
\begin{pmatrix}0&1\\-\omega^2&0\end{pmatrix}\mathbf{x}.
$$

O estado completo do oscilador é $(y,y')$, e não apenas $y$.

## 4. Picard–Lindelöf continua funcionando

Para

$$
\mathbf{x}'=\mathbf F(t,\mathbf{x}),
\qquad
\mathbf{x}(t_0)=\mathbf{x}_0,
$$

a formulação integral é

$$
\boxed{
\mathbf{x}(t)=
\mathbf{x}_0+\int_{t_0}^{t}\mathbf F(s,\mathbf{x}(s))\,ds.
}
$$

Definimos o operador de Picard

$$
(\mathcal K\mathbf{x})(t)
=
\mathbf{x}_0+\int_{t_0}^{t}\mathbf F(s,\mathbf{x}(s))\,ds.
$$

Logo,

$$
\boxed{
\mathbf{x}\text{ resolve o PVI}
\Longleftrightarrow
\mathcal K\mathbf{x}=\mathbf{x}.
}
$$

Agora trabalhamos em $C(I,\mathbb R^n)$ com

$$
\|\mathbf{x}\|_\infty=
\max_{t\in I}\|\mathbf{x}(t)\|.
$$

Esse espaço é Banach.


## 5. Teorema de Picard–Lindelöf para sistemas

Suponha que $\mathbf F$ seja contínua em uma vizinhança de $(t_0,\mathbf{x}_0)$ e localmente Lipschitz em relação a $\mathbf{x}$.

Então existe um intervalo em torno de $t_0$ no qual o PVI possui uma única solução.

A hipótese de Lipschitz é

$$
\|\mathbf F(t,\mathbf{x})-\mathbf F(t,\mathbf{y})\|
\leq L\|\mathbf{x}-\mathbf{y}\|.
$$

Para o operador de Picard,

$$
\begin{aligned}
\|(\mathcal K\mathbf{x})(t)-(\mathcal K\mathbf{y})(t)\|
&\leq
\int_{t_0}^{t}
\|\mathbf F(s,\mathbf{x}(s))-\mathbf F(s,\mathbf{y}(s))\|\,ds\\
&\leq
L|t-t_0|\,\|\mathbf{x}-\mathbf{y}\|_\infty.
\end{aligned}
$$

Assim,

$$
\boxed{
\|\mathcal K\mathbf{x}-\mathcal K\mathbf{y}\|_\infty
\leq Lh\|\mathbf{x}-\mathbf{y}\|_\infty.
}
$$

Se $Lh<1$, $\mathcal K$ é uma contração. A estrutura da prova é a mesma do caso escalar.


## 6. Sistemas autônomos e plano de fase

Um sistema autônomo tem a forma

$$
\boxed{\mathbf{x}'=\mathbf F(\mathbf{x}).}
$$

Em duas dimensões,

$$
\begin{cases}
x'=f(x,y),\\
y'=g(x,y).
\end{cases}
$$

Em cada ponto $(x,y)$, o vetor

$$
\mathbf F(x,y)=
\begin{pmatrix}f(x,y)\\g(x,y)\end{pmatrix}
$$

indica a direção instantânea do movimento.

A solução

$$
t\mapsto(x(t),y(t))
$$

desenha uma **trajetória** no **plano de fase**. O conjunto das trajetórias, com sua orientação temporal, constitui o **retrato de fase**.

**Atenção:** o tempo parametriza a trajetória, mas não é um eixo do plano de fase.


## 7. Primeiro exemplo geométrico

Considere

$$
\begin{cases}
x'=y,\\
y'=-x.
\end{cases}
$$

Sem resolver explicitamente,

$$
\frac{d}{dt}(x^2+y^2)
=2xx'+2yy'
=2xy-2xy=0.
$$

Portanto,

$$
\boxed{x^2+y^2=C.}
$$

As trajetórias não triviais estão contidas em circunferências.

Para determinar a orientação, observe o ponto $(1,0)$:

$$
(x',y')=(0,-1).
$$

O movimento é, portanto, no sentido horário.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 21)
y = np.linspace(-3, 3, 21)
X, Y = np.meshgrid(x, y)
U, V = Y, -X

N = np.sqrt(U**2 + V**2)
N[N == 0] = 1

plt.figure(figsize=(7,7))
plt.quiver(X, Y, U/N, V/N)

theta = np.linspace(0, 2*np.pi, 500)
for r in [0.7, 1.4, 2.1, 2.8]:
    plt.plot(r*np.cos(theta), r*np.sin(theta))

plt.axhline(0, linewidth=.8)
plt.axvline(0, linewidth=.8)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Campo vetorial e trajetórias: x'=y, y'=-x")
plt.axis("equal"); plt.grid(alpha=.25)
plt.show()
```

## 8. Unicidade e geometria

Por que duas trajetórias distintas não podem se cruzar?

Se duas trajetórias passassem pelo mesmo ponto $\mathbf{x}_*$, poderíamos escolher esse instante como $t=0$. Teríamos duas soluções do mesmo PVI

$$
\mathbf{x}'=\mathbf F(\mathbf{x}),
\qquad
\mathbf{x}(0)=\mathbf{x}_*.
$$

Sob as hipóteses de Picard–Lindelöf, isso contradiz a unicidade.

Portanto,

$$
\boxed{
\text{trajetórias distintas de um sistema com unicidade não podem se cruzar.}
}
$$


## 9. Pontos de equilíbrio

Um ponto $\mathbf{x}^*$ é um equilíbrio de

$$
\mathbf{x}'=\mathbf F(\mathbf{x})
$$

quando

$$
\boxed{\mathbf F(\mathbf{x}^*)=\mathbf 0.}
$$

Então

$$
\mathbf{x}(t)\equiv\mathbf{x}^*
$$

é uma solução constante.

No plano, os equilíbrios satisfazem simultaneamente

$$
f(x,y)=0,
\qquad
g(x,y)=0.
$$

### Exemplo

Para

$$
x'=x-y,\qquad y'=x+y,
$$

temos

$$
x-y=0,\qquad x+y=0,
$$

e o único equilíbrio é

$$
\boxed{(0,0).}
$$


## 10. Sistemas lineares

Um sistema linear homogêneo é

$$
\boxed{\mathbf{x}'=A\mathbf{x}},
$$

onde, em duas dimensões,

$$
A=
\begin{pmatrix}
a_{11}&a_{12}\\
a_{21}&a_{22}
\end{pmatrix}.
$$

A matriz $A$ determina o vetor velocidade $A\mathbf{x}$ em cada ponto.

Considere agora

$$
\begin{cases}
x'=x,\\
y'=-y.
\end{cases}
$$

As equações estão desacopladas:

$$
x(t)=C_1e^t,
\qquad
y(t)=C_2e^{-t}.
$$

Logo,

$$
x(t)y(t)=C_1C_2,
$$

e as trajetórias não especiais satisfazem

$$
\boxed{xy=C.}
$$

Os eixos também são trajetórias e devem ser tratados separadamente.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def sistema(t, z):
    x, y = z
    return [x, -y]

grid = np.linspace(-3, 3, 21)
X, Y = np.meshgrid(grid, grid)
U, V = X, -Y
N = np.sqrt(U**2 + V**2)
N[N == 0] = 1

plt.figure(figsize=(7,7))
plt.quiver(X, Y, U/N, V/N)

iniciais = [(1,1),(2,.5),(-1,1),(1,-1),(-1,-1)]
for z0 in iniciais:
    for intervalo in ([0,2],[0,-2]):
        sol = solve_ivp(sistema, intervalo, z0, max_step=.02)
        plt.plot(sol.y[0], sol.y[1])

plt.axhline(0, linewidth=.8); plt.axvline(0, linewidth=.8)
plt.xlim(-3,3); plt.ylim(-3,3)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Retrato de fase: x'=x, y'=-y")
plt.grid(alpha=.25)
plt.show()
```

## 11. Uma primeira ideia de estabilidade em duas dimensões

A origem é equilíbrio de

$$
x'=x,\qquad y'=-y.
$$

Ao longo do eixo $y$,

$$
x(t)=0,\qquad y(t)=C_2e^{-t}\to0.
$$

Ao longo do eixo $x$,

$$
x(t)=C_1e^t,\qquad y(t)=0,
$$

e as soluções afastam-se da origem.

Há, portanto, uma direção de aproximação e outra de afastamento.

Esse equilíbrio é chamado de **sela**.

Mais adiante veremos que essas direções estão associadas aos autovetores de $A$.


## 12. Investigação antes da teoria geral

Considere

$$
\begin{cases}
x'=-2x+y,\\
y'=x-2y.
\end{cases}
$$

Antes de resolver:

1. encontre o equilíbrio;
2. calcule os vetores do campo em $(1,0)$, $(0,1)$, $(1,1)$ e $(1,-1)$;
3. tente prever se soluções próximas da origem se aproximam ou se afastam;
4. procure retas que pareçam preservadas pelo fluxo.

A ideia é procurar **direções especiais** antes de introduzir formalmente autovalores e autovetores.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

A = np.array([[-2.,1.],[1.,-2.]])

def F(t,z):
    return A @ z

g = np.linspace(-3,3,23)
X,Y = np.meshgrid(g,g)
U,V = -2*X+Y, X-2*Y
N = np.sqrt(U**2+V**2)
N[N==0]=1

plt.figure(figsize=(7,7))
plt.quiver(X,Y,U/N,V/N)

for z0 in [(2,0),(0,2),(-2,0),(0,-2),(2,2),(-2,-2),(2,-2),(-2,2),(2,1),(1,2)]:
    sol = solve_ivp(F,[0,4],z0,max_step=.03)
    plt.plot(sol.y[0],sol.y[1])

plt.axhline(0,linewidth=.8); plt.axvline(0,linewidth=.8)
plt.xlim(-3,3); plt.ylim(-3,3)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Investigação: x'=-2x+y, y'=x-2y")
plt.grid(alpha=.25)
plt.show()
```

## 13. A ponte para autovalores e autovetores

O retrato anterior sugere direções especiais nas quais as trajetórias permanecem retas.

Se $\mathbf v$ determina uma dessas direções, precisamos que $A\mathbf v$ seja paralelo a $\mathbf v$. Portanto,

$$
\boxed{A\mathbf v=\lambda\mathbf v.}
$$

Esse é exatamente o problema de autovalores e autovetores da Álgebra Linear.

Na próxima aula veremos como essa estrutura algébrica permite construir e classificar o retrato de fase de sistemas lineares.
