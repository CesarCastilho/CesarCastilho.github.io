# 9. Existência e Unicidade para Problemas de Valor Inicial

## O Teorema de Picard–Lindelöf via o Princípio da Contração

Nesta aula demonstraremos um dos resultados fundamentais da teoria de Equações Diferenciais Ordinárias. Considere

$$
\boxed{\begin{cases}
x'(t)=f(t,x(t)),\\
x(t_0)=x_0,
\end{cases}}
$$

com $x(t)\in\mathbb{R}^n$; o caso escalar corresponde a $n=1$.

O objetivo é entender, por que sob hipóteses adequadas existe uma única solução local. A ideia é transformar o problema de valor inicial (PVI daqui em diante) em um problema de ponto fixo.

$$
\boxed{\text{PVI}\Longleftrightarrow\text{equação integral}\Longleftrightarrow\text{problema de ponto fixo}.}
$$

Para isso ibtroduziremos uma série de conceitos fundamentais: espaços normados, completude, espaços de Banach, norma do supremo, contrações, Teorema do Ponto Fixo de Banach, condição de Lipschitz e iterações de Picard.

# Solução do PVI como um ponto fixo em um espaço de funções.

Integrando formalmente $x'=f(t,x)$ entre $t_0$ e $t$ obtemos,

$$
x(t)=x_0+\int_{t_0}^{t}f(s,x(s))\,ds.
$$

Defina

$$
\boxed{(Kx)(t) \equiv x_0+\int_{t_0}^{t}f(s,x(s))\,ds.}
$$

Resolver o PVI será equivalente a encontrar uma função $x(t)$ tal que

$$
\boxed{K\, x=x.}
$$

O “ponto” fixo é agora uma função. Precisamos agora formalizar o que entendemos por distância, convergência e completude em um conjunto de funções.

# 2 Espaços vetoriais normados


Uma **norma** em um espaço vetorial real $X$ é uma aplicação $\|\cdot\|:X\to[0,\infty)$ tal que

$$
\|x\|=0\Longleftrightarrow x=0,
$$

$$
\|\alpha x\|=|\alpha|\,\|x\|,
$$

$$
\|x+y\|\leq\|x\|+\|y\| \, ,
$$

para todo $ x \in X $ e  todo $\alpha \in \mathbb{R}$.

O par $(X,\|\cdot\|)$ é o que chamamos de um **espaço vetorial normado**.

Em $\mathbb{R}^n$ usaremos a norma euclidiana

$$
\|x \|=\left(\sum_{j=1}^{n}|x_j|^2\right)^{1/2}.
$$

# Desigualdade triangular inversa

Da desigualdade triangular segue

$$
\boxed{|\|x\|-\|y\||\leq\|x-y\|.}
$$

De fato,

$$
\|x\|=\|(x-y)+y\|\leq\|x-y\|+\|y\|
$$

implica $\|x\|-\|y\|\leq\|x-y\|$. Trocando $x$ e $y$ obtemos a desigualdade oposta e, juntando as duas, o resultado.

Uma consequência é que a própria função norma é contínua.

#  Convergência, sequências de Cauchy e completude

Dizemos que $x_n\to x$ se

$$
\|x_n-x\|\to0.
$$

Uma sequência $(x_n)$ é de **Cauchy** se, para todo $\varepsilon>0$, existe $N$ tal que

$$
m,n\geq N\Longrightarrow\|x_n-x_m\|<\varepsilon.
$$

### Exemplos de sequências de Cauchy

#### Exemplo 1 — Uma sequência em $\mathbb{R}$

Considere a sequência

$$
x_n=\frac{1}{n}.
$$

Vamos mostrar diretamente que $(x_n)$ é uma sequência de Cauchy.

Dados $m,n\in\mathbb{N}$, temos

$$
|x_n-x_m|
=
\left|\frac{1}{n}-\frac{1}{m}\right|.
$$

Se $m,n\geq N$, então

$$
\left|\frac{1}{n}-\frac{1}{m}\right|
\leq
\frac{1}{n}+\frac{1}{m}
\leq
\frac{2}{N}.
$$

Dado $\varepsilon>0$, basta escolher $N$ suficientemente grande para que

$$
\frac{2}{N}<\varepsilon.
$$

Por exemplo, podemos tomar

$$
N>\frac{2}{\varepsilon}.
$$

Então, para quaisquer $m,n\geq N$,

$$
|x_n-x_m|<\varepsilon.
$$

Portanto,

$$
\boxed{(x_n)\text{ é uma sequência de Cauchy em }\mathbb{R}.}
$$

Neste caso já sabemos também que

$$
x_n\longrightarrow0.
$$

Observe, entretanto, que a definição de sequência de Cauchy **não exige que conheçamos previamente o limite**.

Ela compara os termos da sequência entre si.

---

#### Exemplo 2 — Uma sequência de funções

Considere agora o espaço

$$
C([0,1])
$$

com a norma do supremo

$$
\|f\|_\infty
=
\max_{x\in[0,1]}|f(x)|.
$$

Defina a sequência de funções

$$
f_n(x)=x+\frac{x}{n}.
$$

Queremos verificar se $(f_n)$ é uma sequência de Cauchy nesse espaço.

Para $m,n\in\mathbb{N}$,

$$
f_n(x)-f_m(x)
=
x\left(\frac{1}{n}-\frac{1}{m}\right).
$$

Portanto,

$$
\begin{aligned}
\|f_n-f_m\|_\infty
&=
\max_{x\in[0,1]}
\left|
x\left(\frac{1}{n}-\frac{1}{m}\right)
\right|\\
&=
\left|\frac{1}{n}-\frac{1}{m}\right|,
\end{aligned}
$$

pois

$$
\max_{x\in[0,1]}|x|=1.
$$

Se $m,n\geq N$, então

$$
\|f_n-f_m\|_\infty
\leq
\frac{2}{N}.
$$

Assim, dado $\varepsilon>0$, escolhendo

$$
N>\frac{2}{\varepsilon},
$$

temos

$$
m,n\geq N
\quad\Longrightarrow\quad
\|f_n-f_m\|_\infty<\varepsilon.
$$

Logo,

$$
\boxed{(f_n)\text{ é uma sequência de Cauchy em }
(C([0,1]),\|\cdot\|_\infty).}
$$

Além disso,

$$
f_n(x)\longrightarrow x,
$$

e, mais precisamente,

$$
\|f_n-f\|_\infty
=
\left\|\frac{x}{n}\right\|_\infty
=
\frac{1}{n}
\longrightarrow0,
$$

onde

$$
f(x)=x.
$$

Portanto, a convergência é uniforme.

## $\color{blue}{\text{Definição — Espaço de Banach}}$


Um espaço normado é **completo** se toda sequência de Cauchy converge para um elemento do próprio espaço.

### Exemplos de espaços normados que não são completos

#### Exemplo 1 — Os números racionais $\mathbb{Q}$

Considere o espaço vetorial

$$
X=\mathbb{Q},
$$

considerado como espaço vetorial sobre o corpo $\mathbb{Q}$, com a norma

$$
\|x\|=|x|.
$$

Esse espaço **não é completo**.

Para ver isso, considere a sequência de números racionais

$$
x_n=
\frac{\lfloor 10^n\sqrt{2}\rfloor}{10^n}.
$$

Os primeiros termos são

$$
1,\quad
1.4,\quad
1.41,\quad
1.414,\quad
1.4142,\quad\ldots
$$

Cada $x_n$ pertence a $\mathbb{Q}$ e

$$
x_n\longrightarrow\sqrt{2}
$$

em $\mathbb{R}$.

Como toda sequência convergente em $\mathbb{R}$ é de Cauchy, $(x_n)$ é
uma sequência de Cauchy. Entretanto,

$$
\sqrt{2}\notin\mathbb{Q}.
$$

Portanto, $(x_n)$ é uma sequência de Cauchy em $\mathbb{Q}$ que não
converge para nenhum elemento de $\mathbb{Q}$.

Assim,

$$
\boxed{
(\mathbb{Q},|\cdot|)\text{ não é completo}.
}
$$


A completude exige que os limites das sequências de Cauchy permaneçam
dentro do próprio espaço.

---

#### Exemplo 2 — O espaço dos polinômios com a norma do supremo

Considere

$$
\mathcal{P}([0,1])
=
\{p:[0,1]\to\mathbb{R}:p\text{ é um polinômio}\},
$$

com a norma

$$
\|p\|_\infty
=
\max_{x\in[0,1]}|p(x)|.
$$

Esse é um espaço vetorial normado, mas **não é completo**.

Considere a sequência de polinômios

$$
p_n(x)
=
\sum_{k=0}^{n}\frac{x^k}{k!}.
$$

Sabemos que

$$
e^x
=
\sum_{k=0}^{\infty}\frac{x^k}{k!}.
$$

Para $x\in[0,1]$,

$$
|e^x-p_n(x)|
=
\left|
\sum_{k=n+1}^{\infty}\frac{x^k}{k!}
\right|
\leq
\sum_{k=n+1}^{\infty}\frac{1}{k!}.
$$

Como

$$
\sum_{k=n+1}^{\infty}\frac{1}{k!}
\longrightarrow0,
$$

temos

$$
\|e^x-p_n\|_\infty\longrightarrow0.
$$

Portanto,

$$
p_n\longrightarrow e^x
$$

uniformemente em $[0,1]$.

Em particular, $(p_n)$ é uma sequência de Cauchy na norma do supremo.

## Exemplo — Uma sequência de Cauchy na norma do máximo

Considere o espaço normado

$$
C([a,b]), \qquad
\|f\|_\infty=\max_{x\in[a,b]}|f(x)|,
$$

e a sequência de polinômios

$$
p_n(x)=\sum_{k=0}^{n}\frac{x^k}{k!}.
$$

Sabemos que esses são os polinômios obtidos truncando a série de Taylor da função exponencial. Vamos mostrar **diretamente**, a partir da definição, que $(p_n)$ é uma sequência de Cauchy em

$$
\left(C([a,b]),\|\cdot\|_\infty\right).
$$

### Demonstração

Sejam $m>n$. Temos

$$
p_m(x)
=
\sum_{k=0}^{m}\frac{x^k}{k!}
$$

e

$$
p_n(x)
=
\sum_{k=0}^{n}\frac{x^k}{k!}.
$$

Subtraindo,

$$
p_m(x)-p_n(x)
=
\sum_{k=n+1}^{m}\frac{x^k}{k!}.
$$

Portanto, pela desigualdade triangular,

$$
|p_m(x)-p_n(x)|
\leq
\sum_{k=n+1}^{m}\frac{|x|^k}{k!}.
$$

Como estamos trabalhando no intervalo compacto $[a,b]$, definimos

$$
M=\max\{|a|,|b|\}.
$$

Assim,

$$
|x|\leq M
\qquad
\text{para todo }x\in[a,b].
$$

Consequentemente,

$$
|p_m(x)-p_n(x)|
\leq
\sum_{k=n+1}^{m}\frac{M^k}{k!}.
$$

Agora tomamos o máximo sobre $x\in[a,b]$. Pela definição da norma do máximo,

$$
\|p_m-p_n\|_\infty
=
\max_{x\in[a,b]}|p_m(x)-p_n(x)|.
$$

Logo,

$$
\|p_m-p_n\|_\infty
\leq
\sum_{k=n+1}^{m}\frac{M^k}{k!}.
$$

Além disso,

$$
\sum_{k=n+1}^{m}\frac{M^k}{k!}
\leq
\sum_{k=n+1}^{\infty}\frac{M^k}{k!}.
$$

Portanto,

$$
\boxed{
\|p_m-p_n\|_\infty
\leq
\sum_{k=n+1}^{\infty}\frac{M^k}{k!}.
}
$$

Mas a série numérica

$$
\sum_{k=0}^{\infty}\frac{M^k}{k!}
$$

é convergente. De fato,

$$
\sum_{k=0}^{\infty}\frac{M^k}{k!}=e^M.
$$

Como toda série convergente possui caudas que tendem a zero,

$$
\sum_{k=n+1}^{\infty}\frac{M^k}{k!}
\longrightarrow 0
\qquad\text{quando }n\to\infty.
$$

Portanto, dado $\varepsilon>0$, existe $N\in\mathbb{N}$ tal que

$$
\sum_{k=N+1}^{\infty}\frac{M^k}{k!}
<\varepsilon.
$$

Se $m,n\geq N$, supondo sem perda de generalidade que $m>n$, temos

$$
\begin{aligned}
\|p_m-p_n\|_\infty
&\leq
\sum_{k=n+1}^{\infty}\frac{M^k}{k!}\\
&\leq
\sum_{k=N+1}^{\infty}\frac{M^k}{k!}\\
&<\varepsilon.
\end{aligned}
$$

Assim, para todo $\varepsilon>0$, existe $N\in\mathbb{N}$ tal que

$$
m,n\geq N
\quad\Longrightarrow\quad
\|p_m-p_n\|_\infty<\varepsilon.
$$

Esta é precisamente a definição de sequência de Cauchy. Portanto,

$$
\boxed{
(p_n)\text{ é uma sequência de Cauchy em }
\left(C([a,b]),\|\cdot\|_\infty\right).
}
$$

---

### Observação conceitual

Observação : **não precisamos conhecer previamente o limite da sequência $(p_n)$ para provar que ela é de Cauchy**.

Comparamos diretamente dois termos da sequência,

$$
p_m
\qquad\text{e}\qquad
p_n,
$$

e mostramos que

$$
\|p_m-p_n\|_\infty\longrightarrow0
\qquad
(m,n\to\infty).
$$

Agora entra a **completude**.

Como

$$
\left(C([a,b]),\|\cdot\|_\infty\right)
$$

é um espaço de Banach, toda sequência de Cauchy nesse espaço converge para algum elemento do próprio espaço. Portanto, existe uma função

$$
f\in C([a,b])
$$

tal que

$$
\|p_n-f\|_\infty\longrightarrow0.
$$

Neste caso particular, sabemos adicionalmente que

$$
f(x)=e^x.
$$


Entretanto,

$$
e^x
$$

não é um polinômio. Logo,

$$
e^x\notin\mathcal{P}([0,1]).
$$

Assim, encontramos uma sequência de Cauchy de elementos de
$\mathcal{P}([0,1])$ cujo limite não pertence a $\mathcal{P}([0,1])$.

Portanto,

$$
\boxed{
\left(\mathcal{P}([0,1]),\|\cdot\|_\infty\right)
\text{ não é completo}.
}
$$


Observe que o espaço

$$
\left(\mathcal{P}([0,1]),\|\cdot\|_\infty\right)
$$

não é completo, enquanto

$$
\boxed{
\left(C([0,1]),\|\cdot\|_\infty\right)
}
$$

é completo e, portanto, é um espaço de Banach.

---

Um espaço vetorial normado completo é chamado de **espaço de Banach**.

$$
\boxed{\text{Banach}=\text{espaço normado completo}.}
$$

# O espaço $C(I,\mathbb{R}^n)$ e a norma do supremo

Seja $I=[a,b]$ e denote por $C(I,\mathbb{R}^n)$ o espaço das funções contínuas $x:I\to\mathbb{R}^n$.

Defina

$$
\boxed{\|x\|_\infty \equiv \sup_{t\in I}|x(t)|=\max_{t\in I}|x(t)|.}
$$

A norma do máximo está bem definica pois o valor máximo existe para cada elemento $x(t)$ dado que  $x(t)$ é uma função contínua e  $I$ ser um conjunto fechado e limitado (compacto).

A distância entre duas funções é

$$
\|x-y\|_\infty=\max_{t\in I}|x(t)-y(t)|.
$$

A convergência nessa norma é exatamente a convergência uniforme.

# A norma do supremo é realmente uma norma

As propriedades são imediatas ponto a ponto:

$$
\|x\|_\infty\geq0,
$$

$$
\|x\|_\infty=0\Longleftrightarrow x(t)=0\ \forall t,
$$

$$
\|\alpha x\|_\infty=|\alpha|\,\|x\|_\infty,
$$

$$
\|x+y\|_\infty\leq\|x\|_\infty+\|y\|_\infty.
$$

# Limite uniforme de funções contínuas é contínuo

## Lema

Se $x_n\in C(I,\mathbb{R}^n)$ e $x_n\to x$ uniformemente, então $x$ é contínua.

### Demonstração

Fixe $t\in I$ e $\varepsilon>0$. Escolha $N$ tal que

$$
\|x_N-x\|_\infty<\frac{\varepsilon}{3}.
$$

Como $x_N$ é contínua, existe $\delta>0$ tal que $|t-s|<\delta$ implica

$$
|x_N(t)-x_N(s)|<\frac{\varepsilon}{3}.
$$

Então

$$
\begin{aligned}
|x(t)-x(s)|
&\leq|x(t)-x_N(t)|+|x_N(t)-x_N(s)|+|x_N(s)-x(s)|\\
&<\varepsilon.
\end{aligned}
$$

Logo $x$ é contínua.

#  Teorema: $C(I,\mathbb{R}^n)$ é Banach

Seja $(x_n)$ Cauchy na norma do supremo. Para cada $t$ fixo,

$$
|x_n(t)-x_m(t)|\leq\|x_n-x_m\|_\infty,
$$

logo $(x_n(t))$ é Cauchy em $\mathbb{R}^n$. Como $\mathbb{R}^n$ é completo, existe

$$
x(t)=\lim_{n\to\infty}x_n(t).
$$

Da propriedade de Cauchy, para $n,m\geq N$,

$$
|x_n(t)-x_m(t)|<\varepsilon
$$

uniformemente em $t$. Fazendo $m\to\infty$,

$$
|x_n(t)-x(t)|\leq\varepsilon
$$

para todo $t$. Assim $x_n\to x$ uniformemente. Pelo lema anterior, $x$ é contínua.

Portanto

$$
\boxed{C(I,\mathbb{R}^n)\text{ é um espaço de Banach}.}
$$

# Subconjuntos fechados de espaços de Banach

## Lema

Se $X$ é Banach e $C\subset X$ é fechado, então $C$ é completo com a métrica induzida.

De fato, uma sequência de Cauchy em $C$ é Cauchy em $X$, logo converge em $X$. Como $C$ é fechado, seu limite pertence a $C$.

Esse detalhe será essencial: o processo iterativo precisa convergir dentro do conjunto onde o operador está definido.

# Contrações

Uma aplicação $K:C\to C$ é uma **contração** se existe $0\leq\theta<1$ tal que

$$
\boxed{\|Kx-Ky\|\leq\theta\|x-y\|}
$$

para todos $x,y\in C$.

Uma contração diminui distâncias por um fator uniforme menor que $1$.

#  Princípio da Contração de Banach

## Teorema

Se $C$ é não vazio e fechado em um espaço de Banach $X$ e $K:C\to C$ é contração, então existe um único $x^*\in C$ com

$$
Kx^*=x^*.
$$

Além disso, para qualquer $x_0\in C$, a sequência $x_{n+1}=Kx_n$ converge para $x^*$ e

$$
\boxed{\|x_n-x^*\|\leq\frac{\theta^n}{1-\theta}\|x_1-x_0\|.}
$$

#  Prova do Princípio da Contração — unicidade

Se $x^*$ e $\widetilde{x}$ são pontos fixos,

$$
\|x^*-\widetilde{x}\|=\|Kx^*-K\widetilde{x}\|\leq\theta\|x^*-\widetilde{x}\|.
$$

Como $\theta<1$,

$$
(1-\theta)\|x^*-\widetilde{x}\|\leq0,
$$

portanto $\|x^*-\widetilde{x}\|=0$ e $x^*=\widetilde{x}$.

#  Prova do Princípio da Contração — existência

Escolha $x_0\in C$ e defina $x_n=K^n(x_0)$. Então

$$
\|x_{n+1}-x_n\|\leq\theta^n\|x_1-x_0\|.
$$

Para $n>m$,

$$
\begin{aligned}
\|x_n-x_m\|
&\leq\sum_{j=m}^{n-1}\|x_{j+1}-x_j\|\\
&\leq\sum_{j=m}^{n-1}\theta^j\|x_1-x_0\|\\
&\leq\frac{\theta^m}{1-\theta}\|x_1-x_0\|.
\end{aligned}
$$

Logo $(x_n)$ é Cauchy. Como $C$ é completo, $x_n\to x^*\in C$.

Como $K$ é Lipschitz, é contínua. Assim

$$
Kx^*=K\left(\lim x_n\right)=\lim Kx_n=\lim x_{n+1}=x^*.
$$

Portanto $x^*$ é ponto fixo. Fazendo $n\to\infty$ na estimativa anterior obtemos a estimativa de erro.

# Onde cada hipótese foi usada?

- **Contração:** produz a estimativa geométrica e mostra que os iterados são Cauchy.
- **Completude:** garante a existência do limite.
- **Fechadura de $C$:** garante que o limite continua pertencendo a $C$.
- **$K:C\to C$:** impede que as iterações abandonem o conjunto.

Esse mapa lógico é importante para compreender a demonstração, e não apenas reproduzi-la.

#  Equivalência entre PVI e equação integral

## Lema

Uma função contínua $x$ satisfaz

$$
x(t)=x_0+\int_{t_0}^{t}f(s,x(s))\,ds
$$

se e somente se é uma solução $C^1$ do PVI

$$
x'=f(t,x),\qquad x(t_0)=x_0.
$$

### PVI $\Rightarrow$ equação integral

Pelo Teorema Fundamental do Cálculo,

$$
x(t)-x(t_0)=\int_{t_0}^{t}x'(s)\,ds=\int_{t_0}^{t}f(s,x(s))\,ds.
$$

### Equação integral $\Rightarrow$ PVI

Se $x$ é contínua e $f$ é contínua, então $s\mapsto f(s,x(s))$ é contínua. Logo o Teorema Fundamental do Cálculo dá

$$
x'(t)=f(t,x(t)).
$$

Tomando $t=t_0$ na equação integral, obtemos $x(t_0)=x_0$.

#  O operador de Picard e as aproximações sucessivas

Defina

$$
\boxed{(Kx)(t)=x_0+\int_{t_0}^{t}f(s,x(s))\,ds.}
$$

Começando pela função constante $x_0(t)\equiv x_0$,

$$
x_1=Kx_0,
$$

$$
x_2=Kx_1,
$$

$$
\cdots
$$

ou, explicitamente,

$$
\boxed{x_{m+1}(t)=x_0+\int_{t_0}^{t}f(s,x_m(s))\,ds.}
$$

Essas são as **iterações de Picard**.

#  Um exemplo concreto

Considere o problema de valor inicial

$$
y'=2t(1+y),\qquad y(0)=0.
$$

Integrando de $0$ até $t$, obtemos a equação integral equivalente

$$
\varphi(t)=\int_0^t 2s\bigl(1+\varphi(s)\bigr)\,ds.
$$

Escolhemos inicialmente

$$
\varphi_0(t)=0
$$

e definimos recursivamente

$$
\boxed{\displaystyle
\varphi_{n+1}(t)=\int_0^t2s\bigl(1+\varphi_n(s)\bigr)\,ds.
}
$$

Essa é a **iteração de Picard**. Observe que estamos iterando **funções**.

A sequência

$$
\varphi_0,\varphi_1,\varphi_2,\ldots
$$

deve ser pensada como uma sequência de pontos em um espaço de funções. É precisamente por isso que introduzimos espaços normados, completude e espaços de Banach antes de enunciar o argumento geral.

# Calculando as primeiras aproximações

A primeira aproximação é

$$
\begin{aligned}
\varphi_1(t)
&=\int_0^t2s\bigl(1+\varphi_0(s)\bigr)\,ds\\
&=\int_0^t2s\,ds\\
&=t^2.
\end{aligned}
$$

Para a segunda,

$$
\begin{aligned}
\varphi_2(t)
&=\int_0^t2s\bigl(1+\varphi_1(s)\bigr)\,ds\\
&=\int_0^t2s(1+s^2)\,ds\\
&=t^2+\frac{t^4}{2}.
\end{aligned}
$$

E para a terceira,

$$
\begin{aligned}
\varphi_3(t)
&=\int_0^t2s\left(1+s^2+\frac{s^4}{2}\right)\,ds\\
&=t^2+\frac{t^4}{2}+\frac{t^6}{2\cdot3}.
\end{aligned}
$$

Isso sugere

$$
\boxed{\displaystyle
\varphi_n(t)
=t^2+\frac{t^4}{2!}+\frac{t^6}{3!}
+\cdots+\frac{t^{2n}}{n!}.
}
$$

### Verificação por indução

Suponha que

$$
\varphi_k(t)=\sum_{j=1}^{k}\frac{t^{2j}}{j!}.
$$

Então

$$
\begin{aligned}
\varphi_{k+1}(t)
&=\int_0^t2s\left(1+\sum_{j=1}^{k}\frac{s^{2j}}{j!}\right)\,ds\\
&=\int_0^t\left(2s+\sum_{j=1}^{k}\frac{2s^{2j+1}}{j!}\right)\,ds\\
&=t^2+\sum_{j=1}^{k}\frac{t^{2j+2}}{(j+1)!}\\
&=\sum_{j=1}^{k+1}\frac{t^{2j}}{j!}.
\end{aligned}
$$

Portanto a fórmula vale para todo $n\geq1$.

# O que as iterações estão fazendo geometricamente?

A figura abaixo, extraída da referência, mostra as primeiras aproximações de Picard para esse PVI.

Observe um fenômeno importante: conforme $n$ aumenta, as curvas $\varphi_n$ permanecem próximas umas das outras em intervalos progressivamente maiores.

Visualmente, isso sugere convergência. Mas **uma figura não é uma demonstração de convergência**. Nossa tarefa matemática será transformar essa observação em estimativas capazes de garantir convergência no espaço de funções.

Essa distinção é importante:

- o gráfico **sugere** o fenômeno;
- a norma do supremo permite **medir** a distância entre funções;
- a completude permite concluir que uma sequência de Cauchy possui limite;
- o argumento de contração garante que o limite é um ponto fixo do operador.

# Identificando o limite

Temos

$$
\varphi_n(t)=\sum_{k=1}^{n}\frac{t^{2k}}{k!}.
$$

Para $t$ fixo, o teste da razão fornece

$$
\left|
\frac{t^{2(k+1)}/(k+1)!}{t^{2k}/k!}
\right|
=
\frac{t^2}{k+1}
\longrightarrow0.
$$

Logo a série converge para todo $t\in\mathbb{R}$ e

$$
\varphi(t)=\sum_{k=1}^{\infty}\frac{t^{2k}}{k!}.
$$

Reconhecendo a série exponencial,

$$
e^{t^2}
=
\sum_{k=0}^{\infty}\frac{t^{2k}}{k!},
$$

obtemos

$$
\boxed{\varphi(t)=e^{t^2}-1.}
$$

Neste conseguimos identificar explicitamente a função limite. **A prova do teorema de existência não pode depender disso.** Em geral, não saberemos escrever a solução em forma fechada.

O objetivo do teorema será provar que o limite existe e satisfaz o PVI mesmo quando não conseguimos encontrá-lo explicitamente.

picard_erros.png## 9.13E O erro das aproximações

Como neste exemplo conhecemos a solução exata,

$$
\varphi(t)=e^{t^2}-1,
$$

podemos definir

$$
e_n(t)=\varphi(t)-\varphi_n(t).
$$

A figura da referência mostra $e_n(t)$ para as primeiras iterações.



Novamente aparece uma ideia que será central na prova: para controlar uma função em **todo um intervalo**, não basta observar seu valor em um ponto. Queremos controlar uma quantidade do tipo

$$
\|e_n\|_\infty
=
\max_{|t|\leq T}|e_n(t)|.
$$

É exatamente essa necessidade que explica a escolha da norma do supremo em $C(I,\mathbb{R}^n)$.

## As dificuldades da demonstração

 Ao definirmos

$$
\varphi_{n+1}=K\varphi_n,
$$

precisamos responder:

1. **Existência das iteradas:** $K\varphi_n$ está definido para todo $n$?
2. **Convergência:** a sequência $(\varphi_n)$ converge?
3. **Identificação do limite:** se $\varphi_n\to\varphi$, podemos concluir que
   $$
   \varphi=K\varphi?
   $$
4. **Unicidade:** pode existir outro ponto fixo de $K$?

Abordaremos essas questões na demonstração de Picard--Lindelöf:

- $K(C)\subset C$ garante que a iteração não sai do conjunto onde estamos trabalhando;
- a contração transforma as iteradas em uma sequência de Cauchy;
- a completude garante a existência do limite;
- a continuidade de $K$ permite identificar o limite como ponto fixo;
- a propriedade de contração fornece unicidade.

#  A restrição do intervalo

Daremos agora uma interpretação geométrica especialmente útil para a escolha do intervalo de existência.

Suponha, após uma translação do ponto inicial, que

$$
y'=f(t,y),\qquad y(0)=0,
$$

e que $f$ seja contínua no retângulo

$$
R=\{(t,y):|t|\leq a,\ |y|\leq b\}.
$$

Como $R$ é compacto e $f$ é contínua, existe $M>0$ tal que

$$
|f(t,y)|\leq M
$$

em $R$.

Mas, para cada iterada,

$$
\varphi_{n+1}'(t)=f(t,\varphi_n(t)).
$$

Portanto,

$$
|\varphi_{n+1}'(t)|\leq M.
$$

Como $\varphi_{n+1}(0)=0$, pelo Teorema Fundamental do Cálculo

$$
|\varphi_{n+1}(t)|
=
\left|\int_0^t\varphi_{n+1}'(s)\,ds\right|
\leq M|t|.
$$

Assim, todas as iteradas ficam confinadas à região

$$
|y|\leq M|t|.
$$

Geometricamente, essa é a região em forma de **gravata-borboleta** mostrada abaixo.
picard_retangulo.png

Para que essa região permaneça dentro de $|y|\leq b$, precisamos de

$$
M|t|\leq b,
$$

isto é,

$$
|t|\leq\frac{b}{M}.
$$

Como também precisamos permanecer em $|t|\leq a$, tomamos

$$
\boxed{\displaystyle h=\min\left\{a,\frac{b}{M}\right\}.}
$$

Esta figura torna muito concreta uma das duas razões pelas quais reduzimos o intervalo de tempo: **precisamos impedir que as iterações saiam da região onde conhecemos $f$.**

Mais adiante aparece uma segunda restrição, de natureza diferente: tornar o operador uma contração. Não devemos confundir essas duas ideias.

#  A região onde trabalharemos

Como $U$ é aberto e $(t_0,x_0)\in U$, podemos escolher $T>0$ e $\delta>0$ tais que

$$
V=[t_0-T,t_0+T]\times\overline{B_\delta(x_0)}\subset U.
$$

Como $V$ é compacto e $f$ é contínua,

$$
\boxed{M=\max_{(t,x)\in V}|f(t,x)|<\infty.}
$$

O número $M$ controla o tamanho máximo do campo vetorial dentro da região.

#  Por que aparece $\delta/M$?

Se $x$ permanece na bola $|x(t)-x_0|\leq\delta$, então

$$
\begin{aligned}
|(Kx)(t)-x_0|
&\leq\int_{t_0}^{t}|f(s,x(s))|\,|ds|\\
&\leq M|t-t_0|.
\end{aligned}
$$

Assim, se

$$
|t-t_0|\leq\frac{\delta}{M},
$$

temos

$$
|(Kx)(t)-x_0|\leq\delta.
$$

Portanto escolhemos

$$
\boxed{T_0\leq\min\left\{T,\frac{\delta}{M}\right\}.}
$$

Se $M=0$, interpretamos $\delta/M=\infty$.

#  Outra rota para enxergar a convergência

O argumento pelo Princípio da Contração é conceitualmente muito poderoso. A referência também mostra uma rota mais concreta, que ajuda a entender de onde surgem as estimativas.

Escrevemos

$$
\varphi_n
=
\varphi_1+
(\varphi_2-\varphi_1)+
\cdots+
(\varphi_n-\varphi_{n-1}).
$$

Assim, estudar a convergência de $(\varphi_n)$ equivale a controlar a série de funções formada pelos incrementos

$$
\varphi_{k+1}-\varphi_k.
$$

Se $f$ satisfaz uma condição de Lipschitz

$$
|f(t,y_1)-f(t,y_2)|
\leq K|y_1-y_2|,
$$

e $|f(t,y)|\leq M$, obtém-se sucessivamente

$$
|\varphi_1(t)|\leq M|t|,
$$

$$
|\varphi_2(t)-\varphi_1(t)|
\leq
\frac{MK|t|^2}{2!},
$$

e, por indução,

$$
\boxed{\displaystyle
|\varphi_n(t)-\varphi_{n-1}(t)|
\leq
\frac{M K^{n-1}|t|^n}{n!}.
}
$$

Em $|t|\leq h$,

$$
|\varphi_n(t)-\varphi_{n-1}(t)|
\leq
\frac{M K^{n-1}h^n}{n!}.
$$

A série numérica correspondente converge, pois é dominada por uma série exponencial. Pelo teste de Weierstrass, a série de funções converge uniformemente.

Essa abordagem é muito instrutiva porque exibe explicitamente o aparecimento do fatorial $n!$. Ela também antecipa a ideia usada posteriormente no Teorema de Weissinger.

#  O espaço completo onde o operador atuará

Defina

$$
I_0=[t_0-T_0,t_0+T_0]
$$

e

$$
X=C(I_0,\mathbb{R}^n),
$$

com norma do supremo. Dentro de $X$, considere

$$
\boxed{C=\{x\in X:\|x-x_0\|_\infty\leq\delta\},}
$$

onde $x_0$ representa a função constante $t\mapsto x_0$.

$X$ é Banach e $C$ é fechado; portanto $C$ é completo.

#  Prova de que $K(C)\subset C$

Se $x\in C$, seu gráfico permanece em $V$. Para $t\in I_0$,

$$
|(Kx)(t)-x_0|\leq M|t-t_0|\leq MT_0\leq\delta.
$$

Logo

$$
\|Kx-x_0\|_\infty\leq\delta,
$$

isto é,

$$
\boxed{K(C)\subset C.}
$$

Esta é a primeira razão para escolher o intervalo de tempo pequeno.

#  Lipschitz

A continuidade de $f$ não fornece, sozinha, um controle linear da diferença $f(t,x)-f(t,y)$. Para isso usamos a condição de Lipschitz.

Dizemos que $f$ é **Lipschitz em $x$, uniformemente em $t$**, em uma região $V$, se existe $L\geq0$ tal que

$$
\boxed{|f(t,x)-f(t,y)|\leq L|x-y|}
$$

para todos $(t,x),(t,y)\in V$.

Dizemos que é **localmente Lipschitz** quando tal constante existe em cada subconjunto compacto da região.

#  Prova de que $K$ é contração

Para $x,y\in C$,

$$
\begin{aligned}
|(Kx)(t)-(Ky)(t)|
&\leq\int_{t_0}^{t}|f(s,x(s))-f(s,y(s))|\,|ds|\\
&\leq L\int_{t_0}^{t}|x(s)-y(s)|\,|ds|\\
&\leq L|t-t_0|\,\|x-y\|_\infty\\
&\leq LT_0\|x-y\|_\infty.
\end{aligned}
$$

Tomando o máximo,

$$
\boxed{\|Kx-Ky\|_\infty\leq LT_0\|x-y\|_\infty.}
$$

Se

$$
\boxed{LT_0<1,}
$$

então $K$ é uma contração.

#  Por que convergência uniforme aparece naturalmente?

Há duas passagens que seriam perigosas se tivéssemos apenas convergência pontual.

Primeiro, uma sequência de funções contínuas pode convergir pontualmente para uma função descontínua. Um exemplo clássico é

$$
\varphi_n(x)=x^n,\qquad 0\leq x\leq1.
$$

Temos

$$
\varphi_n(x)\longrightarrow
\begin{cases}
0,&0\leq x<1,\\
1,&x=1,
\end{cases}
$$

e o limite não é contínuo.

Segundo, em geral não podemos simplesmente escrever

$$
\lim_{n\to\infty}\int_a^b\varphi_n(s)\,ds
=
\int_a^b\lim_{n\to\infty}\varphi_n(s)\,ds.
$$

Por isso a norma do supremo e a convergência uniforme  são o mecanismo que permite controlar simultaneamente todos os pontos do intervalo e justificar a passagem ao limite na equação integral.

#  Duas razões distintas para diminuir $T_0$

É importante não confundir as duas estimativas:

$$
T_0\leq\frac{\delta}{M}
$$

garante

$$
K(C)\subset C,
$$

enquanto

$$
T_0<\frac1L
$$

garante que $K$ seja uma contração.

São papéis matemáticos diferentes.

# Teorema de Picard–Lindelöf

## Teorema

Seja $f\in C(U,\mathbb{R}^n)$, com $U\subset\mathbb{R}\times\mathbb{R}^n$ aberto e $(t_0,x_0)\in U$. Suponha que $f$ seja localmente Lipschitz na segunda variável, uniformemente na primeira.

Então existe um intervalo $I$ contendo $t_0$ no qual

$$
\begin{cases}
x'=f(t,x),\\
x(t_0)=x_0
\end{cases}
$$

possui uma única solução local $x\in C^1(I,\mathbb{R}^n)$.

#  Demonstração de Picard–Lindelöf

Escolha $V=[t_0-T,t_0+T]\times\overline{B_\delta(x_0)}\subset U$. Pela continuidade,

$$
M=\max_V|f|<\infty.
$$

Pela hipótese de Lipschitz local, existe $L$ com

$$
|f(t,x)-f(t,y)|\leq L|x-y|
$$

em $V$.

Escolha $T_0>0$ tal que

$$
T_0\leq T,\qquad MT_0\leq\delta,\qquad LT_0<1.
$$

Em

$$
X=C([t_0-T_0,t_0+T_0],\mathbb{R}^n)
$$

considere a bola fechada

$$
C=\{x:\|x-x_0\|_\infty\leq\delta\}.
$$

Já provamos que $C$ é completo, que $K(C)\subset C$ e que $K$ é contração. Pelo Teorema do Ponto Fixo de Banach, existe um único $x\in C$ com $Kx=x$.

Pela equivalência entre equação integral e PVI, essa função é exatamente a única solução local.

$$
\boxed{\exists!\,x(t).}
$$

#  O resultado é local, não necessariamente global

Considere

$$
x'=x^2,\qquad x(0)=1.
$$

A solução explícita é

$$
x(t)=\frac{1}{1-t}.
$$

Ela explode em $t=1$. Portanto, Picard–Lindelöf garante existência e unicidade **local**, mas não afirma que a solução exista para todo $t\in\mathbb{R}$.

```python
import numpy as np
import matplotlib.pyplot as plt

t=np.linspace(-2,.98,600)
x=1/(1-t)
plt.figure(figsize=(8,5))
plt.plot(t,x)
plt.axvline(1,linestyle="--",linewidth=1)
plt.xlabel("t"); plt.ylabel("x(t)")
plt.title(r"Solução de $x'=x^2,\;x(0)=1$")
plt.ylim(-1,30); plt.grid(alpha=.25)
plt.show()
```

#  Continuidade não garante unicidade

Considere

$$
x'=\sqrt{|x|},\qquad x(0)=0.
$$

$f(x)=\sqrt{|x|}$ é contínua, mas não é Lipschitz em $0$, pois

$$
\frac{|f(x)-f(0)|}{|x|}=\frac1{\sqrt{|x|}}\to\infty.
$$

Além da solução $x(t)\equiv0$, para cada $a\geq0$ temos

$$
x_a(t)=\begin{cases}
0,&t\leq a,\\[1mm]
\dfrac{(t-a)^2}{4},&t\geq a,
\end{cases}
$$

que também resolve o PVI em $t\geq0$.

Logo continuidade de $f$ pode garantir existência em resultados mais gerais, mas não unicidade.

#  Por que $C^1$ implica Lipschitz local — caso escalar

Se $f_y$ é contínua em uma região compacta, então é limitada:

$$
|f_y(t,y)|\leq L.
$$

Fixando $t$, o Teorema do Valor Médio dá

$$
f(t,y_1)-f(t,y_2)=f_y(t,\xi)(y_1-y_2)
$$

para algum $\xi$ entre $y_1$ e $y_2$. Portanto

$$
\boxed{|f(t,y_1)-f(t,y_2)|\leq L|y_1-y_2|.}
$$

Assim, a condição prática

$$
f,\ f_y\text{ contínuas perto de }(t_0,y_0)
$$

é suficiente para existência e unicidade local.

#  Versão vetorial: o Jacobiano

Para $f:\mathbb{R}^m\to\mathbb{R}^n$, considere o segmento

$$
\gamma(\varepsilon)=x+\varepsilon(y-x),\qquad0\leq\varepsilon\leq1.
$$

Pelo Teorema Fundamental do Cálculo,

$$
f(y)-f(x)=\int_0^1Df(\gamma(\varepsilon))(y-x)\,d\varepsilon.
$$

Logo

$$
\boxed{|f(y)-f(x)|\leq\left(\sup_{0\leq\varepsilon\leq1}\|Df(\gamma(\varepsilon))\|\right)|y-x|.}
$$

Se $Df$ é contínua, é limitada em compactos. Portanto

$$
\boxed{f\in C^1\Longrightarrow f\text{ é localmente Lipschitz}.}
$$

#  Três exemplos técnicos perto de $0$

### (a) $f(x)=\dfrac{1}{1-x^2}$

Em qualquer intervalo compacto contido em $(-1,1)$, $f'$ é contínua e limitada. Logo $f$ é Lipschitz perto de $0$.

### (b) $f(x)=|x|^{1/2}$

Não é Lipschitz em $0$, pois

$$
\frac{|f(x)-f(0)|}{|x|}=\frac1{\sqrt{|x|}}\to\infty.
$$

### (c) $f(x)=x^2\sin(1/x)$ para $x\neq0$, $f(0)=0$

Para $x\neq0$,

$$
f'(x)=2x\sin(1/x)-\cos(1/x),
$$

que é limitada perto de $0$. Além disso,

$$
f'(0)=\lim_{h\to0}h\sin(1/h)=0.
$$

Logo, pelo Teorema do Valor Médio, $f$ é Lipschitz perto de $0$, embora $f'$ não seja contínua em $0$.

Isso mostra que $f_y$ contínua é condição suficiente, mas não necessária, para Lipschitz local.

# Regularidade adicional da solução

Se $f\in C^k$, $k\geq1$, então a solução local pertence a $C^{k+1}$.

De fato, Picard–Lindelöf fornece $x\in C^1$. Como

$$
x'(t)=f(t,x(t)),
$$

se $f\in C^1$, o lado direito é $C^1$, logo $x'\in C^1$ e $x\in C^2$. O argumento prossegue por indução.

$$
\boxed{f\in C^k\Longrightarrow x\in C^{k+1}.}
$$

#  Exemplo clássico das iterações de Picard: $x'=x$

Considere

$$
x'=x,\qquad x(0)=1.
$$

A equação integral é

$$
x(t)=1+\int_0^t x(s)\,ds.
$$

Começando com $x_0(t)=1$,

$$
x_1(t)=1+t,
$$

$$
x_2(t)=1+t+\frac{t^2}{2!},
$$

$$
x_3(t)=1+t+\frac{t^2}{2!}+\frac{t^3}{3!}.
$$

Por indução,

$$
\boxed{x_n(t)=\sum_{j=0}^{n}\frac{t^j}{j!}.}
$$

Logo $x_n\to e^t$.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial

t=np.linspace(-1.5,1.5,500)
plt.figure(figsize=(9,5))
for n in [0,1,2,3,5,8]:
    p=np.zeros_like(t)
    for j in range(n+1):
        p += t**j/factorial(j)
    plt.plot(t,p,label=fr"$x_{n}$")
plt.plot(t,np.exp(t),"--",linewidth=2,label=r"$e^t$")
plt.xlabel("t"); plt.ylabel("x")
plt.title(r"Iterações de Picard para $x'=x,\;x(0)=1$")
plt.grid(alpha=.25); plt.legend(); plt.show()
```

#  Um exemplo mais delicado

Considere

$$
x'=2t-2\sqrt{\max(0,x)},\qquad x(0)=0.
$$

A função é contínua, mas não é Lipschitz em $x$ perto de $0$. A iteração de Picard ainda pode ser escrita formalmente como

$$
x_{n+1}(t)=\int_0^t\left[2s-2\sqrt{\max(0,x_n(s))}\right]ds.
$$

A ausência da hipótese de Lipschitz indica que o teorema não fornece unicidade; esse é precisamente o tipo de situação em que precisamos examinar o problema com mais cuidado.

#  Uma sutileza: a condição $LT_0<1$ não é intrínseca

A prova mais simples usa $LT_0<1$. Entretanto, para as iteradas do operador de Picard vale a estimativa mais forte

$$
\boxed{\|K^n x-K^n y\|_\infty\leq\frac{(LT_0)^n}{n!}\|x-y\|_\infty.}
$$

Ela segue por indução. Se a estimativa vale para $n$, então

$$
\begin{aligned}
|(K^{n+1}x)(t)-(K^{n+1}y)(t)|
&\leq L\int_{t_0}^{t}|(K^n x)(s)-(K^n y)(s)|\,ds\\
&\leq\frac{(L|t-t_0|)^{n+1}}{(n+1)!}\|x-y\|_\infty.
\end{aligned}
$$

Como

$$
\frac{(LT_0)^n}{n!}\to0,
$$

alguma iterada $K^n$ é uma contração, mesmo se $LT_0\geq1$.

#  Teorema de Weissinger — extensão do princípio da contração

Uma generalização útil afirma: se $C$ é fechado em um Banach e

$$
\|K^n x-K^n y\|\leq\theta_n\|x-y\|
$$

com

$$
\sum_{n=1}^{\infty}\theta_n<\infty,
$$

então $K$ possui um único ponto fixo.

Para Picard,

$$
\theta_n=\frac{(LT_0)^n}{n!},
$$

logo

$$
\sum_{n=1}^{\infty}\theta_n=e^{LT_0}-1<\infty.
$$

Assim, a restrição $LT_0<1$ pode ser removida da formulação final do teorema. O controle $T_0\leq\delta/M$, porém, continua sendo útil para manter a solução dentro da região onde as hipóteses foram verificadas.

#  Exemplo completo de aplicação do teorema

Considere

$$
y'=x+y^2,\qquad y(0)=0.
$$

Escolha

$$
|x|\leq\frac12,\qquad |y|\leq1.
$$

Então

$$
|f(x,y)|\leq\frac12+1=\frac32,
$$

logo podemos tomar $M=3/2$.

Além disso,

$$
f_y(x,y)=2y,
$$

portanto $|f_y|\leq2$ e podemos tomar $L=2$.

Com $\delta=1$, a condição de invariância pede

$$
T_0\leq\frac{2}{3},
$$

e a prova direta por contração pede

$$
T_0<\frac12.
$$

Assim, $T_0=1/4$ é uma escolha simples. O teorema garante uma única solução em pelo menos

$$
\left[-\frac14,\frac14\right].
$$

Esse intervalo é uma garantia do teorema, não necessariamente o intervalo máximo da solução.

#  Exemplo: equação linear de primeira ordem

Para

$$
y'=a(t)y+b(t),
$$

com $a,b$ contínuas,

$$
f(t,y)=a(t)y+b(t).
$$

Então

$$
|f(t,y_1)-f(t,y_2)|=|a(t)|\,|y_1-y_2|.
$$

Em qualquer intervalo compacto, $|a(t)|$ é limitada. Logo $f$ é localmente Lipschitz em $y$.

Assim, todo PVI linear de primeira ordem com coeficientes contínuos possui solução local única, em concordância com a fórmula obtida anteriormente pelo fator integrante.

#  Exemplo: critério prático com $f_y$

Considere

$$
y'=\sin(xy)+e^x,\qquad y(0)=1.
$$

Temos

$$
f(x,y)=\sin(xy)+e^x,
$$

$$
f_y(x,y)=x\cos(xy).
$$

Ambas são contínuas em todo o plano. Portanto $f$ é localmente Lipschitz em $y$ em toda região compacta e, por Picard–Lindelöf, existe uma única solução local passando por qualquer ponto inicial.

#  Curvas integrais distintas não podem se cruzar

Suponha que duas soluções $x_1$ e $x_2$ de

$$
x'=f(t,x)
$$

satisfaçam

$$
x_1(t_*)=x_2(t_*).
$$

Então ambas resolvem o mesmo PVI com condição inicial em $t_*$. Pela unicidade local, elas coincidem numa vizinhança de $t_*$. Portanto, em uma região onde vale Picard–Lindelöf, trajetórias distintas não podem cruzar-se.

#  A demonstração

$$
\boxed{\text{PVI}}
\Longleftrightarrow
\boxed{\text{equação integral}}
\Longleftrightarrow
\boxed{Kx=x}.
$$

Depois:

$$
f\text{ contínua}\Longrightarrow M<\infty\Longrightarrow K(C)\subset C,
$$

$$
f\text{ Lipschitz em }x\Longrightarrow\|Kx-Ky\|\leq LT_0\|x-y\|.
$$

Logo, para $T_0$ adequado, $K$ é contração e Banach fornece

$$
\boxed{\exists!\,x:Kx=x.}
$$

Finalmente, esse ponto fixo é exatamente a solução do PVI.

# Resumo

O Teorema de Picard–Lindelöf nasce da combinação de quatro ideias:

$$
\boxed{C(I,\mathbb{R}^n)\text{ é Banach}},
$$

$$
\boxed{\text{PVI}\Longleftrightarrow Kx=x},
$$

$$
\boxed{K(C)\subset C},
$$

$$
\boxed{K\text{ é contração}}.
$$

O Teorema do Ponto Fixo de Banach fornece então um único ponto fixo do operador integral, e esse ponto fixo é precisamente a solução do problema de valor inicial.

A mensagem conceitual central é:

$$
\boxed{\text{existência e unicidade de uma EDO}\Longleftrightarrow\text{existência e unicidade de um ponto fixo adequado}.}
$$
